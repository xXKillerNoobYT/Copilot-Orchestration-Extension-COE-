# GitHub Issues Sync Integration

**Extension Dependency**: `hiroyannnn.vscode-github-issues-sync`  
**Marketplace**: https://marketplace.visualstudio.com/items?itemName=hiroyannnn.vscode-github-issues-sync  
**Purpose**: Synchronize ZenTasks with GitHub Issues automatically

---

## Setup Instructions

### 1. Install GitHub Issues Sync Extension

Install the extension dependency:
```bash
code --install-extension hiroyannnn.vscode-github-issues-sync
```

Or manually in VS Code:
1. Open Extensions (Ctrl+Shift+X)
2. Search: `hiroyannnn.vscode-github-issues-sync`
3. Click **Install**

### 2. Configure Extension

Add to `.vscode/settings.json`:
```json
{
  "githubIssuesSync.enabled": true,
  "githubIssuesSync.repository": "xXKillerNoobYT/Copilot-Orchestration-Extension-COE-",
  "githubIssuesSync.syncDirection": "bidirectional",
  "githubIssuesSync.syncInterval": 300000,
  "githubIssuesSync.localDirectory": ".github/issues",
  "githubIssuesSync.autoSync": true,
  "githubIssuesSync.syncOnStartup": true,
  "githubIssuesSync.createMissingLabels": true,
  "githubIssuesSync.taskLabelMapping": {
    "feature": "type:feature",
    "bug": "type:bug",
    "refactor": "type:refactor",
    "testing": "type:testing",
    "documentation": "type:documentation",
    "critical": "priority:critical",
    "high": "priority:high",
    "medium": "priority:medium",
    "low": "priority:low"
  }
}
```

### 3. Authentication

Generate GitHub Personal Access Token:
1. Go to https://github.com/settings/tokens
2. Generate new token (classic)
3. Select scopes: `repo`, `workflow`
4. Copy token

Add to VS Code settings:
```bash
# Command Palette (Ctrl+Shift+P)
> GitHub Issues Sync: Set GitHub Token
# Paste token when prompted
```

### 4. Folder Structure

Issues will sync to:
```
.github/
  issues/
    issue-1.md
    issue-2.md
    issue-3.md
    ...
```

### 5. Task Migration Process

**Automatic Migration:**
```bash
# Command Palette
> GitHub Issues Sync: Migrate Local Tasks to GitHub
```

**Manual Migration:**
1. Open each task in `_ZENTASKS/`
2. Extract YAML frontmatter
3. Create GitHub issue via API
4. Map task fields to issue fields:
   - `title` → Issue title
   - `description` → Issue body
   - `priority` → Label
   - `task_type` → Label
   - `dependencies` → Issue references in body
   - `status` → Issue state (open/closed)

### 6. Verify Sync

Check sync status:
```bash
# Command Palette
> GitHub Issues Sync: Show Sync Status
```

Verify all issues created:
```bash
# Visit repository
https://github.com/xXKillerNoobYT/Copilot-Orchestration-Extension-COE-/issues
```

## Overview

This notebook documents the complete planning phase for **Planner Mode**—a comprehensive UI/UX system within the VS Code extension that allows users to create, visualize, and manage development plans before any code is written.

**Key Goal**: Get accurate planning specifications before implementation begins.

### What is Planner Mode?

Planner Mode transforms raw requirements into structured, validated development plans with:
- ✓ Task decomposition and hierarchy
- ✓ Dependency mapping and critical path analysis
- ✓ Timeline and resource allocation
- ✓ AI-powered suggestions and validation
- ✓ Multi-format export (JSON, Markdown, CSV, PDF, GitHub Issues)
- ✓ Template library for common workflows

## Section 1: Define Project Objectives

### Planning Phase Objectives

The Planner Mode must achieve these core objectives:

````python
import json
from dataclasses import dataclass
from typing import List, Dict
from enum import Enum

# Define planning phase objectives using structured data

@dataclass
class ProjectObjective:
    """Structured representation of a planning objective"""
    id: str
    name: str
    description: str
    measurable_outcome: str
    priority: str  # critical, high, medium, low
    success_criteria: List[str]

# Core objectives for Planner Mode
objectives = [
    ProjectObjective(
        id="obj-001",
        name="Intuitive Requirement Capture",
        description="Allow users to input raw requirements in free-form text",
        measurable_outcome="Users can create a plan from unstructured requirements in <5 minutes",
        priority="critical",
        success_criteria=[
            "Text area accepts free-form input",
            "AI parses and suggests task breakdown",
            "Users can refine suggestions easily"
        ]
    ),
    ProjectObjective(
        id="obj-002",
        name="Atomic Task Decomposition",
        description="Break down large epics into manageable, completable tasks",
        measurable_outcome="All tasks can be completed in 1-4 hours",
        priority="critical",
        success_criteria=[
            "Hierarchical task editor (epics → stories → subtasks)",
            "Drag-and-drop reorganization",
            "Validation that tasks are <4 hours each"
        ]
    ),
    ProjectObjective(
        id="obj-003",
        name="Dependency Mapping & Critical Path",
        description="Visualize task relationships and identify critical sequences",
        measurable_outcome="Visual graph shows all dependencies with no circular references",
        priority="high",
        success_criteria=[
            "Network diagram visualization",
            "Hard/soft dependency types",
            "Auto-detection of circular dependencies",
            "Critical path highlighting"
        ]
    ),
    ProjectObjective(
        id="obj-004",
        name="Timeline & Resource Planning",
        description="Calculate project timeline and allocate team resources",
        measurable_outcome="Gantt chart shows realistic timeline based on dependencies",
        priority="high",
        success_criteria=[
            "Timeline auto-calculated from dependencies",
            "Resource allocation view",
            "Capacity planning alerts"
        ]
    ),
    ProjectObjective(
        id="obj-005",
        name="Plan Validation & Quality Gates",
        description="Ensure plans are complete, realistic, and ready to execute",
        measurable_outcome="Validation checklist shows all quality gates met",
        priority="high",
        success_criteria=[
            "Circular dependency detection",
            "Orphan task identification",
            "Acceptance criteria verification",
            "Realism checks (no 100+ hour tasks)"
        ]
    ),
    ProjectObjective(
        id="obj-006",
        name="Multi-Format Export & Integration",
        description="Export plans to various formats and GitHub Issues",
        measurable_outcome="Plans can be exported to JSON, Markdown, CSV, PDF, and GitHub",
        priority="medium",
        success_criteria=[
            "JSON export (full data)",
            "Markdown export (documentation)",
            "CSV export (spreadsheet)",
            "PDF export (printable report)",
            "GitHub Issues creation"
        ]
    ),
    ProjectObjective(
        id="obj-007",
        name="Template-Driven Planning",
        description="Provide pre-built templates to accelerate planning for common scenarios",
        measurable_outcome="Users can select template and customize in <10 minutes",
        priority="medium",
        success_criteria=[
            "Feature Implementation template",
            "Bug Fix Sprint template",
            "Refactor template",
            "Custom template creation"
        ]
    ),
    ProjectObjective(
        id="obj-008",
        name="AI-Powered Assistance",
        description="Use AI to suggest tasks, dependencies, estimates, and risks",
        measurable_outcome="AI suggestions improve plan quality and reduce creation time",
        priority="high",
        success_criteria=[
            "Task decomposition suggestions",
            "Dependency inference",
            "Effort estimation",
            "Risk identification",
            "Acceptance criteria generation"
        ]
    )
]

# Display objectives
print("=" * 80)
print("PLANNER MODE - PROJECT OBJECTIVES")
print("=" * 80)
print()

for obj in objectives:
    print(f"[{obj.priority.upper()}] {obj.id}: {obj.name}")
    print(f"  Description: {obj.description}")
    print(f"  Measurable Outcome: {obj.measurable_outcome}")
    print(f"  Success Criteria:")
    for criterion in obj.success_criteria:
        print(f"    ✓ {criterion}")
    print()

# Summary statistics
critical_count = sum(1 for o in objectives if o.priority == "critical")
high_count = sum(1 for o in objectives if o.priority == "high")
medium_count = sum(1 for o in objectives if o.priority == "medium")

print("=" * 80)
print(f"SUMMARY: {critical_count} Critical | {high_count} High | {medium_count} Medium priorities")
print("=" * 80)
````

## Section 2: Identify Stakeholders and Requirements

### Stakeholder Analysis

Who uses Planner Mode and what do they need?

````python
import pandas as pd
import matplotlib.pyplot as plt
from dataclasses import dataclass
from typing import List

@dataclass
class Stakeholder:
    """Stakeholder profile with influence and interest assessment"""
    name: str
    role: str
    influence: int  # 1-10 scale
    interest: int   # 1-10 scale
    needs: List[str]
    requirements: List[str]

# Define stakeholder matrix
stakeholders = [
    Stakeholder(
        name="Project Manager / Tech Lead",
        role="Plan Creator & Approver",
        influence=9,
        interest=10,
        needs=[
            "Quick plan creation from requirements",
            "Dependency visibility for critical path",
            "Resource allocation tools",
            "Plan validation before handoff"
        ],
        requirements=[
            "Wizard-based plan creation (<10 min)",
            "Visual dependency graph",
            "Timeline with milestones",
            "Export to GitHub/spreadsheet"
        ]
    ),
    Stakeholder(
        name="Developer",
        role="Task Executor",
        influence=8,
        interest=8,
        needs=[
            "Clear, decomposed tasks",
            "Understand task dependencies",
            "Realistic estimates",
            "Actionable acceptance criteria"
        ],
        requirements=[
            "Tasks <4 hours each",
            "Clear AC for each task",
            "No surprises or missing scope",
            "Predictable execution"
        ]
    ),
    Stakeholder(
        name="QA/Tester",
        role="Plan Validator",
        influence=6,
        interest=7,
        needs=[
            "Test coverage in plan",
            "Clear acceptance criteria",
            "Risk visibility"
        ],
        requirements=[
            "Testing phase explicitly planned",
            "AC written for all tasks",
            "Risk assessment included"
        ]
    ),
    Stakeholder(
        name="Product Owner",
        role="Stakeholder & Approver",
        influence=10,
        interest=9,
        needs=[
            "Understand feature scope",
            "Timeline for release",
            "Risk assessment",
            "GO/NO-GO decision support"
        ],
        requirements=[
            "Executive summary view",
            "Timeline with confidence",
            "Risk & mitigation strategies",
            "Scope clarity (in vs out)"
        ]
    ),
    Stakeholder(
        name="AI/Copilot System",
        role="Plan Assistant",
        influence=7,
        interest=10,
        needs=[
            "Structured requirements input",
            "Clear task definitions",
            "Execution context",
            "Feedback on execution"
        ],
        requirements=[
            "Machine-readable task format",
            "Dependency DAG for sequencing",
            "Acceptance criteria for validation",
            "Integration with GitHub"
        ]
    )
]

# Create stakeholder matrix for visualization
stakeholder_data = {
    'Stakeholder': [s.name for s in stakeholders],
    'Role': [s.role for s in stakeholders],
    'Influence': [s.influence for s in stakeholders],
    'Interest': [s.interest for s in stakeholders]
}

df_stakeholders = pd.DataFrame(stakeholder_data)

# Create Power/Interest matrix visualization
fig, ax = plt.subplots(figsize=(10, 8))

# Plot stakeholders
colors = ['#DC2626' if x >= 7 and y >= 7 else '#EA580C' if x >= 7 or y >= 7 else '#3B82F6' 
          for x, y in zip(df_stakeholders['Influence'], df_stakeholders['Interest'])]

ax.scatter(df_stakeholders['Influence'], df_stakeholders['Interest'], s=500, alpha=0.6, c=colors)

# Add labels
for idx, row in df_stakeholders.iterrows():
    ax.annotate(row['Stakeholder'], 
                (row['Influence'], row['Interest']),
                fontsize=9, ha='center', va='center', weight='bold')

# Add quadrant lines
ax.axhline(y=5, color='gray', linestyle='--', alpha=0.5)
ax.axvline(x=5, color='gray', linestyle='--', alpha=0.5)

# Quadrant labels
ax.text(2.5, 9, 'Keep Satisfied\n(Low Influence,\nHigh Interest)', 
        ha='center', fontsize=9, style='italic', alpha=0.7)
ax.text(8.5, 9, 'Manage Closely\n(High Influence,\nHigh Interest)', 
        ha='center', fontsize=9, style='italic', alpha=0.7, weight='bold')
ax.text(2.5, 1.5, 'Monitor\n(Low Influence,\nLow Interest)', 
        ha='center', fontsize=9, style='italic', alpha=0.7)
ax.text(8.5, 1.5, 'Keep Informed\n(High Influence,\nLow Interest)', 
        ha='center', fontsize=9, style='italic', alpha=0.7)

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_xlabel('Influence', fontsize=12, weight='bold')
ax.set_ylabel('Interest', fontsize=12, weight='bold')
ax.set_title('Stakeholder Matrix - Planner Mode', fontsize=14, weight='bold')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Print stakeholder details
print("\n" + "=" * 100)
print("DETAILED STAKEHOLDER REQUIREMENTS")
print("=" * 100)

for s in stakeholders:
    print(f"\n{s.role.upper()}: {s.name}")
    print(f"Influence: {s.influence}/10 | Interest: {s.interest}/10")
    print(f"\nNeeds:")
    for need in s.needs:
        print(f"  • {need}")
    print(f"\nRequirements for Planner Mode:")
    for req in s.requirements:
        print(f"  ✓ {req}")
    print("-" * 100)
````

## Section 3: Create Project Timeline and Milestones

### Planner Mode Development Timeline

Planning the implementation roadmap with Gantt chart visualization.

````python
%pip install plotly
from dataclasses import dataclass
from typing import List
from datetime import datetime, timedelta
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots

@dataclass
class Phase:
    """Implementation phase with timeline and dependencies"""
    id: str
    name: str
    description: str
    start_date: str
    duration_days: int
    dependencies: List[str]
    deliverables: List[str]
    critical: bool

# Define implementation phases
phases = [
    Phase(
        id="phase-001",
        name="Planning & Design",
        description="Complete UI/UX design, wireframes, and architecture documentation",
        start_date="2026-01-08",
        duration_days=10,
        dependencies=[],
        deliverables=[
            "10 page wireframes/designs",
            "Component library spec",
            "Data model documentation",
            "User workflow documentation",
            "Design system (colors, typography, spacing)"
        ],
        critical=True
    ),
    Phase(
        id="phase-002",
        name="Component Library",
        description="Build reusable Vue components for Planner UI",
        start_date="2026-01-18",
        duration_days=12,
        dependencies=["phase-001"],
        deliverables=[
            "TaskCard component",
            "DependencyGraph component",
            "Timeline component",
            "Modal/Dialog components",
            "Badge/Badge components",
            "FormInputs components"
        ],
        critical=True
    ),
    Phase(
        id="phase-003",
        name="Backend APIs",
        description="Implement Laravel APIs for plan CRUD and validation",
        start_date="2026-01-18",
        duration_days=14,
        dependencies=["phase-001"],
        deliverables=[
            "POST /plans (create plan)",
            "GET /plans/{id} (read plan)",
            "PUT /plans/{id} (update plan)",
            "DELETE /plans/{id} (delete plan)",
            "POST /plans/{id}/validate (validation)",
            "POST /plans/{id}/export (export formats)"
        ],
        critical=True
    ),
    Phase(
        id="phase-004",
        name="Page Implementation",
        description="Build all 10 pages with components and API integration",
        start_date="2026-02-01",
        duration_days=18,
        dependencies=["phase-002", "phase-003"],
        deliverables=[
            "Welcome/Home page",
            "New Plan Wizard",
            "Requirements page",
            "Task Decomposition page",
            "Dependency Graph page",
            "Timeline page",
            "Task Details page",
            "Template Library page",
            "Review & Validation page",
            "Export page"
        ],
        critical=True
    ),
    Phase(
        id="phase-005",
        name="AI Integration",
        description="Integrate with AI services for suggestions and automation",
        start_date="2026-02-19",
        duration_days=10,
        dependencies=["phase-004"],
        deliverables=[
            "Requirements parsing service",
            "Task decomposition suggestions",
            "Dependency inference",
            "Effort estimation",
            "Risk assessment",
            "Acceptance criteria generation"
        ],
        critical=False
    ),
    Phase(
        id="phase-006",
        name="Testing & QA",
        description="Unit tests, integration tests, and user acceptance testing",
        start_date="2026-03-01",
        duration_days=10,
        dependencies=["phase-004"],
        deliverables=[
            "Unit test suite (80%+ coverage)",
            "Integration tests",
            "E2E tests",
            "UAT testing results",
            "Bug fixes and polish"
        ],
        critical=True
    ),
    Phase(
        id="phase-007",
        name="Documentation & Launch",
        description="User guide, API documentation, and release",
        start_date="2026-03-11",
        duration_days=5,
        dependencies=["phase-006"],
        deliverables=[
            "User guide (Markdown)",
            "API documentation",
            "Video tutorials",
            "Release notes",
            "Launch announcement"
        ],
        critical=False
    )
]

# Create Gantt chart data
gantt_data = []
start_base = datetime.strptime(phases[0].start_date, "%Y-%m-%d")

for phase in phases:
    start = datetime.strptime(phase.start_date, "%Y-%m-%d")
    end = start + timedelta(days=phase.duration_days)
    
    gantt_data.append({
        'Task': phase.name,
        'Start': start,
        'End': end,
        'Duration': phase.duration_days,
        'Critical': phase.critical
    })

df_gantt = pd.DataFrame(gantt_data)

# Create Gantt chart with Plotly
fig = go.Figure()

for idx, row in df_gantt.iterrows():
    color = '#DC2626' if row['Critical'] else '#3B82F6'
    fig.add_trace(go.Bar(
        x=[row['Duration']],
        y=[row['Task']],
        orientation='h',
        base=row['Start'],
        marker=dict(color=color),
        text=f"{row['Duration']} days",
        textposition="auto",
        hovertemplate=f"<b>{row['Task']}</b><br>Start: {row['Start'].strftime('%b %d')}<br>End: {row['End'].strftime('%b %d')}<br>Duration: {row['Duration']} days<extra></extra>"
    ))

fig.update_layout(
    title="Planner Mode - Implementation Timeline",
    xaxis_type='date',
    yaxis_title="Phase",
    xaxis_title="Timeline",
    hovermode='closest',
    height=600,
    showlegend=False,
    template='plotly_white'
)

fig.show()

# Print detailed timeline
print("\n" + "=" * 120)
print("IMPLEMENTATION TIMELINE - DETAILED BREAKDOWN")
print("=" * 120)

total_days = sum(p.duration_days for p in phases)
critical_path_days = sum(p.duration_days for p in phases if p.critical)

print(f"\nTotal Timeline: {total_days} calendar days (≈ {total_days // 7} weeks)")
print(f"Critical Path: {critical_path_days} days")

for i, phase in enumerate(phases, 1):
    start = datetime.strptime(phase.start_date, "%Y-%m-%d")
    end = start + timedelta(days=phase.duration_days)
    deps = ", ".join(phase.dependencies) if phase.dependencies else "None"
    critical_badge = "🔴 CRITICAL" if phase.critical else "🔵 Standard"
    
    print(f"\n[{i}] {critical_badge} | {phase.name}")
    print(f"    Timeline: {start.strftime('%b %d')} → {end.strftime('%b %d')} ({phase.duration_days} days)")
    print(f"    Description: {phase.description}")
    print(f"    Dependencies: {deps}")
    print(f"    Deliverables:")
    for deliverable in phase.deliverables:
        print(f"      ✓ {deliverable}")

print("\n" + "=" * 120)
print(f"MILESTONE CHECKPOINTS:")
print("=" * 120)
print(f"  ✓ Jan 18: Planning & Design Complete → Ready for Component/API Build")
print(f"  ✓ Feb 01: Components & APIs Ready → Start Full Page Implementation")
print(f"  ✓ Mar 01: Core Features Complete → Begin Testing & Polish")
print(f"  ✓ Mar 16: Release Ready → Launch Planner Mode")
````

````python
@dataclass
class ResourceAllocation:
    """Resource and effort estimation"""
    role: str
    person_count: int
    hours_per_week: int
    total_weeks: int
    phase_focus: str
    key_responsibilities: List[str]

resources = [
    ResourceAllocation(
        role="UX/Product Designer",
        person_count=1,
        hours_per_week=40,
        total_weeks=4,
        phase_focus="Planning & Design (Phase 1-2)",
        key_responsibilities=[
            "Wireframe all 10 pages",
            "Create high-fidelity designs",
            "Define component library",
            "Design system (colors, typography, spacing)",
            "User flow documentation"
        ]
    ),
    ResourceAllocation(
        role="Frontend Developer (Vue/TypeScript)",
        person_count=1,
        hours_per_week=40,
        total_weeks=6,
        phase_focus="Components & Pages (Phase 2, 4)",
        key_responsibilities=[
            "Build component library",
            "Implement all 10 pages",
            "State management setup",
            "API integration",
            "Frontend testing"
        ]
    ),
    ResourceAllocation(
        role="Backend Developer (Laravel/PHP)",
        person_count=1,
        hours_per_week=40,
        total_weeks=5,
        phase_focus="APIs & Services (Phase 3, 5)",
        key_responsibilities=[
            "REST API implementation",
            "Data models & migrations",
            "Plan validation logic",
            "Export functionality",
            "AI integration endpoints"
        ]
    ),
    ResourceAllocation(
        role="AI/ML Engineer",
        person_count=1,
        hours_per_week=20,
        total_weeks=3,
        phase_focus="AI Integration (Phase 5)",
        key_responsibilities=[
            "Requirements parsing",
            "Task decomposition suggestions",
            "Dependency inference",
            "Effort estimation model",
            "Risk assessment logic"
        ]
    ),
    ResourceAllocation(
        role="QA/Tester",
        person_count=1,
        hours_per_week=30,
        total_weeks=4,
        phase_focus="Testing (Phase 6)",
        key_responsibilities=[
            "Unit test writing",
            "Integration testing",
            "UAT coordination",
            "Bug tracking & validation",
            "Performance testing"
        ]
    ),
    ResourceAllocation(
        role="Tech Lead / PM",
        person_count=1,
        hours_per_week=20,
        total_weeks=7,
        phase_focus="All phases (coordination, review)",
        key_responsibilities=[
            "Architecture oversight",
            "Code reviews",
            "Dependency management",
            "Stakeholder communication",
            "Timeline tracking"
        ]
    )
]

# Create resource allocation visualization
resource_data = []
for r in resources:
    total_hours = r.hours_per_week * r.total_weeks
    resource_data.append({
        'Role': r.role,
        'Count': r.person_count,
        'Hours/Week': r.hours_per_week,
        'Weeks': r.total_weeks,
        'Total Hours': total_hours
    })

df_resources = pd.DataFrame(resource_data)

# Visualization 1: Resource hours allocation
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# Hours by role
ax1.barh(df_resources['Role'], df_resources['Total Hours'], color='#3B82F6')
ax1.set_xlabel('Total Hours', fontweight='bold')
ax1.set_title('Resource Allocation by Role', fontweight='bold', fontsize=12)
ax1.grid(axis='x', alpha=0.3)

# Add value labels
for i, v in enumerate(df_resources['Total Hours']):
    ax1.text(v + 5, i, str(v), va='center')

# Hours per week utilization
colors_util = ['#10B981' if x <= 40 else '#EA580C' if x <= 50 else '#DC2626' 
               for x in df_resources['Hours/Week']]
ax2.barh(df_resources['Role'], df_resources['Hours/Week'], color=colors_util)
ax2.set_xlabel('Hours per Week', fontweight='bold')
ax2.set_title('Weekly Utilization per Role', fontweight='bold', fontsize=12)
ax2.axvline(x=40, color='green', linestyle='--', label='Full-time', alpha=0.7)
ax2.legend()
ax2.grid(axis='x', alpha=0.3)

plt.tight_layout()
plt.show()

# Print resource details
print("\n" + "=" * 140)
print("RESOURCE ALLOCATION AND EFFORT ESTIMATION")
print("=" * 140)

total_person_weeks = sum(r.person_count * r.total_weeks for r in resources)
total_hours = df_resources['Total Hours'].sum()
avg_utilization = total_hours / total_person_weeks / 40 * 100 if total_person_weeks > 0 else 0

print(f"\nTotal Person-Weeks: {total_person_weeks}")
print(f"Total Estimated Hours: {total_hours}")
print(f"Average Utilization: {avg_utilization:.1f}%")

print("\nDETAILED BREAKDOWN:\n")

for r in resources:
    total_hours = r.hours_per_week * r.total_weeks
    print(f"👤 {r.role.upper()}")
    print(f"   Team Size: {r.person_count} person(s)")
    print(f"   Availability: {r.hours_per_week} hours/week × {r.total_weeks} weeks = {total_hours} hours")
    print(f"   Phase Focus: {r.phase_focus}")
    print(f"   Key Responsibilities:")
    for resp in r.key_responsibilities:
        print(f"     • {resp}")
    print()

print("=" * 140)
print("CRITICAL RESOURCES:")
print("=" * 140)
print("  1. Frontend Developer - Cannot be split across multiple people (component consistency)")
print("  2. Backend Developer - Owns all API contracts (must coordinate with Frontend)")
print("  3. UX Designer - Critical path blocker (all pages need design before implementation)")
print("  4. Tech Lead - Must review all architectural decisions")

print("\n" + "=" * 140)
print("RISK FACTORS:")
print("=" * 140)
print("  ⚠️  Single designer could create bottleneck if iterations needed")
print("  ⚠️  Frontend + Backend sync critical (API contract ownership)")
print("  ⚠️  AI engineer part-time (suggest 3 weeks full-time instead of part-time)")
print("  ⚠️  QA late in cycle (recommend concurrent testing during development)")
````

## Section 5: Risk Assessment and Mitigation Strategies

### Planner Mode Implementation Risks

````python
@dataclass
class Risk:
    """Risk identification and mitigation"""
    id: str
    title: str
    description: str
    probability: int  # 1-10
    impact: int       # 1-10
    category: str
    mitigation_strategy: str
    contingency: str
    owner: str

risks = [
    Risk(
        id="risk-001",
        title="Scope Creep - Feature Requests",
        description="Stakeholders request additional features during implementation (templates, advanced analytics, etc.)",
        probability=8,
        impact=8,
        category="Scope",
        mitigation_strategy="Define clear Phase 1 scope. Document all requests as Phase 2 candidates. Hold bi-weekly scope reviews.",
        contingency="Establish change control process. Require priority trade-offs for any new features.",
        owner="Tech Lead"
    ),
    Risk(
        id="risk-002",
        title="UI Complexity - Too Many Features on Pages",
        description="Pages become overcrowded, reducing UX quality and increasing cognitive load",
        probability=6,
        impact=7,
        category="Design/UX",
        mitigation_strategy="Enforce minimalist design principles. Use progressive disclosure (tabs, expandable sections). Conduct UX testing early.",
        contingency="Redesign individual pages, split complex pages into tabs, defer secondary features.",
        owner="UX Designer"
    ),
    Risk(
        id="risk-003",
        title="AI Integration Delays",
        description="AI services (requirements parsing, dependency inference) underperform or require more tuning than expected",
        probability=6,
        impact=6,
        category="Technology",
        mitigation_strategy="Define AI as Phase 5 (not blocking MVP). Plan fallback: manual task input. Start AI POC early.",
        contingency="Deploy MVP without AI suggestions. AI becomes Phase 6 enhancement.",
        owner="AI Engineer"
    ),
    Risk(
        id="risk-004",
        title="Circular Dependency Detection Edge Cases",
        description="Complex task graphs expose bugs in circular dependency detection algorithm",
        probability=5,
        impact=8,
        category="Technical",
        mitigation_strategy="Design algorithm carefully. Test with synthetic complex graphs (>50 tasks). Code review by 2+ engineers.",
        contingency="Ship with known limitation (max 100 tasks). Add async detection if needed.",
        owner="Backend Developer"
    ),
    Risk(
        id="risk-005",
        title="Frontend-Backend API Contract Misalignment",
        description="Frontend and backend teams design APIs independently, causing integration issues",
        probability=7,
        impact=7,
        category="Integration",
        mitigation_strategy="Define API contracts in Phase 1. Hold joint architecture session. Use API mocking early. Frequent sync meetings.",
        contingency="Implement adapter layer to map between API versions. Delay feature until aligned.",
        owner="Tech Lead"
    ),
    Risk(
        id="risk-006",
        title="Performance: Large Task Graphs (>500 tasks)",
        description="Dependency graph visualization and timeline rendering become slow for large plans",
        probability=4,
        impact=9,
        category="Performance",
        mitigation_strategy="Implement lazy loading for large graphs. Profile early. Virtual scrolling for task lists. Pagination for export.",
        contingency="Limit graph visualization to 200 tasks. Implement progressive enhancement.",
        owner="Frontend Developer"
    ),
    Risk(
        id="risk-007",
        title="Resource Unavailability - Key Person Dependency",
        description="Designer or backend lead becomes unavailable (vacation, illness, departure)",
        probability=3,
        impact=9,
        category="Resource",
        mitigation_strategy="Document designs and architecture thoroughly. Pair programming for critical components. Cross-train team members.",
        contingency="Hire contract resources. Pause development until resource available.",
        owner="Tech Lead"
    ),
    Risk(
        id="risk-008",
        title="GitHub Integration Issues",
        description="GitHub API changes, authentication issues, or rate limiting affect issue creation/sync",
        probability=4,
        impact=6,
        category="Integration",
        mitigation_strategy="Abstract GitHub integration behind service layer. Handle rate limiting gracefully. Test with real GitHub API.",
        contingency="Queue issues for batch creation. Provide manual export fallback.",
        owner="Backend Developer"
    ),
    Risk(
        id="risk-009",
        title="Testing Coverage Gaps",
        description="Insufficient test coverage leads to bugs in production (especially edge cases in dependency logic)",
        probability=7,
        impact=8,
        category="Quality",
        mitigation_strategy="Mandate 80%+ code coverage. Test circular dependency detection thoroughly. Pair testing with development.",
        contingency="Extend testing phase. Hire QA consultant for spec-based testing.",
        owner="QA Engineer"
    ),
    Risk(
        id="risk-010",
        title="Adoption Resistance - Learning Curve",
        description="Users find Planner Mode complex and revert to ad-hoc planning",
        probability=5,
        impact=7,
        category="Adoption",
        mitigation_strategy="Create tutorial videos. Build guided first-run experience. Provide template library. Offer training sessions.",
        contingency="Simplify MVP (defer advanced features). Provide direct support to early adopters.",
        owner="Product Manager"
    )
]

# Calculate risk scores
risk_data = []
for r in risks:
    risk_score = r.probability * r.impact
    risk_data.append({
        'ID': r.id,
        'Title': r.title,
        'Probability': r.probability,
        'Impact': r.impact,
        'Risk Score': risk_score,
        'Category': r.category
    })

df_risks = pd.DataFrame(risk_data)
df_risks = df_risks.sort_values('Risk Score', ascending=False)

# Create risk matrix visualization
fig, ax = plt.subplots(figsize=(12, 8))

# Color mapping based on risk score
def get_risk_color(score):
    if score >= 60:
        return '#DC2626'  # Red - Critical
    elif score >= 40:
        return '#EA580C'  # Orange - High
    elif score >= 20:
        return '#FBBF24'  # Amber - Medium
    else:
        return '#3B82F6'  # Blue - Low

colors = [get_risk_color(score) for score in df_risks['Risk Score']]

scatter = ax.scatter(df_risks['Probability'], df_risks['Impact'], 
                     s=[score * 15 for score in df_risks['Risk Score']], 
                     c=colors, alpha=0.6, edgecolors='black', linewidth=1.5)

# Add labels
for idx, row in df_risks.iterrows():
    ax.annotate(row['ID'].replace('risk-', 'R'), 
                (row['Probability'], row['Impact']),
                fontsize=10, ha='center', va='center', weight='bold')

# Risk zones
ax.axhspan(0, 3, alpha=0.05, color='green')
ax.axhspan(3, 6, alpha=0.05, color='yellow')
ax.axhspan(6, 10, alpha=0.05, color='red')

ax.axvspan(0, 3, alpha=0.05, color='green')
ax.axvspan(3, 6, alpha=0.05, color='yellow')
ax.axvspan(6, 10, alpha=0.05, color='red')

ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.set_xlabel('Probability (1-10)', fontweight='bold', fontsize=12)
ax.set_ylabel('Impact (1-10)', fontweight='bold', fontsize=12)
ax.set_title('Risk Assessment Matrix - Planner Mode Implementation', fontweight='bold', fontsize=14)
ax.grid(True, alpha=0.3)

# Add quadrant labels
ax.text(1.5, 8.5, 'Monitor', ha='center', fontsize=10, style='italic', alpha=0.7)
ax.text(8, 8.5, 'CRITICAL\nRISK', ha='center', fontsize=10, style='italic', alpha=0.7, weight='bold')
ax.text(1.5, 1.5, 'Low Risk', ha='center', fontsize=10, style='italic', alpha=0.7)
ax.text(8, 1.5, 'High Impact\nLow Probability', ha='center', fontsize=10, style='italic', alpha=0.7)

plt.tight_layout()
plt.show()

# Print risk details
print("\n" + "=" * 150)
print("DETAILED RISK ASSESSMENT")
print("=" * 150)

print("\nRisk Priority (by Risk Score):\n")
for idx, row in df_risks.iterrows():
    risk = next(r for r in risks if r.id == row['ID'])
    risk_level = "🔴 CRITICAL" if row['Risk Score'] >= 60 else "🟠 HIGH" if row['Risk Score'] >= 40 else "🟡 MEDIUM" if row['Risk Score'] >= 20 else "🔵 LOW"
    
    print(f"{risk_level} | {risk.id}: {risk.title}")
    print(f"    Score: {row['Risk Score']}/100 (P:{row['Probability']} × I:{row['Impact']})")
    print(f"    Description: {risk.description}")
    print(f"    Mitigation: {risk.mitigation_strategy}")
    print(f"    Contingency: {risk.contingency}")
    print(f"    Owner: {risk.owner}")
    print()

print("=" * 150)
print("CRITICAL RISKS REQUIRING IMMEDIATE ACTION:")
print("=" * 150)
critical_risks = [r for r in df_risks[df_risks['Risk Score'] >= 60].itertuples()]
for i, risk_row in enumerate(critical_risks, 1):
    risk = next(r for r in risks if r.id == risk_row.ID)
    print(f"\n{i}. {risk.title}")
    print(f"   Action: {risk.mitigation_strategy}")
````

## Section 6: Define Success Metrics and KPIs

````python
@dataclass
class SuccessMetric:
    """KPI and success measurement"""
    category: str
    metric_name: str
    target: str
    measurement_method: str
    baseline: str
    owner: str
    frequency: str

success_metrics = [
    # Quality Metrics
    SuccessMetric(
        category="Quality",
        metric_name="Code Coverage",
        target="≥80% unit test coverage",
        measurement_method="Coverage reports from CI/CD pipeline",
        baseline="0% (new code)",
        owner="QA Engineer",
        frequency="Every PR"
    ),
    SuccessMetric(
        category="Quality",
        metric_name="Critical Bugs in Production",
        target="Zero critical bugs in first 30 days",
        measurement_method="Issue tracker classification",
        baseline="N/A (beta launch)",
        owner="QA Engineer",
        frequency="Daily"
    ),
    SuccessMetric(
        category="Quality",
        metric_name="Page Load Performance",
        target="All pages load <2 seconds (Lighthouse score >85)",
        measurement_method="Lighthouse CI, Real User Monitoring",
        baseline="N/A (new feature)",
        owner="Frontend Developer",
        frequency="Monthly"
    ),
    
    # Adoption Metrics
    SuccessMetric(
        category="Adoption",
        metric_name="User Activation Rate",
        target="≥60% of users create first plan within 7 days",
        measurement_method="Analytics tracking (GA, Sentry, custom events)",
        baseline="TBD at launch",
        owner="Product Manager",
        frequency="Weekly"
    ),
    SuccessMetric(
        category="Adoption",
        metric_name="Plan Completion Rate",
        target="≥70% of started plans reach 'Ready to Execute' status",
        measurement_method="Plan status tracking in backend",
        baseline="TBD at launch",
        owner="Product Manager",
        frequency="Weekly"
    ),
    SuccessMetric(
        category="Adoption",
        metric_name="Monthly Active Users",
        target="≥100 MAU in first month, growing 15% month-over-month",
        measurement_method="Unique user counts from analytics",
        baseline="0 (new feature)",
        owner="Product Manager",
        frequency="Monthly"
    ),
    
    # Functionality Metrics
    SuccessMetric(
        category="Functionality",
        metric_name="Average Plan Creation Time",
        target="<15 minutes from requirements to 'Ready to Execute'",
        measurement_method="Session duration tracking, user surveys",
        baseline="60+ minutes (manual planning)",
        owner="UX Designer",
        frequency="Quarterly"
    ),
    SuccessMetric(
        category="Functionality",
        metric_name="AI Suggestion Accuracy",
        target="≥75% of AI suggestions accepted by users without modification",
        measurement_method="Suggestion acceptance tracking",
        baseline="TBD (depends on AI model)",
        owner="AI Engineer",
        frequency="Monthly"
    ),
    SuccessMetric(
        category="Functionality",
        metric_name="Circular Dependency Detection",
        target="100% detection rate on test suites",
        measurement_method="Automated tests, manual validation on complex graphs",
        baseline="TBD (algorithm testing)",
        owner="Backend Developer",
        frequency="Per release"
    ),
    
    # User Satisfaction
    SuccessMetric(
        category="Satisfaction",
        metric_name="User Satisfaction (NPS)",
        target="NPS ≥50 (Very Satisfied)",
        measurement_method="In-app surveys, post-plan feedback forms",
        baseline="TBD at launch",
        owner="Product Manager",
        frequency="Monthly"
    ),
    SuccessMetric(
        category="Satisfaction",
        metric_name="Feature Usefulness Rating",
        target="≥4/5 average rating for 'Helps me plan better'",
        measurement_method="Likert scale survey after plan completion",
        baseline="TBD at launch",
        owner="UX Designer",
        frequency="Quarterly"
    ),
    SuccessMetric(
        category="Satisfaction",
        metric_name="Support Ticket Volume",
        target="<1 support ticket per 50 users per month",
        measurement_method="Support system tracking",
        baseline="TBD (depends on launch volume)",
        owner="Tech Lead",
        frequency="Weekly"
    ),
    
    # Business Impact
    SuccessMetric(
        category="Business",
        metric_name="Plan-to-Execution Success Rate",
        target="≥85% of plans executed successfully (no major delays/failures)",
        measurement_method="Task completion tracking, execution feedback",
        baseline="60-70% (without structured plans)",
        owner="Product Manager",
        frequency="Quarterly"
    ),
    SuccessMetric(
        category="Business",
        metric_name="Time Saved per Project",
        target="Average 10+ hours saved per project (planning + execution)",
        measurement_method="User surveys, internal metrics tracking",
        baseline="TBD at launch",
        owner="Product Manager",
        frequency="Quarterly"
    ),
    SuccessMetric(
        category="Business",
        metric_name="Feature Adoption vs. Alternatives",
        target="Planner Mode used for ≥80% of COE-based projects",
        measurement_method="Usage analytics",
        baseline="0% (new feature)",
        owner="Product Manager",
        frequency="Monthly"
    )
]

# Create metrics dashboard
metrics_by_category = {}
for metric in success_metrics:
    if metric.category not in metrics_by_category:
        metrics_by_category[metric.category] = []
    metrics_by_category[metric.category].append(metric)

# Visualization: Metrics by Category
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Success Metrics Dashboard - Planner Mode', fontsize=16, fontweight='bold')

categories = list(metrics_by_category.keys())
metric_counts = [len(metrics_by_category[cat]) for cat in categories]

# Plot 1: Metrics by Category
ax = axes[0, 0]
colors_cat = ['#10B981', '#3B82F6', '#8B5CF6', '#EA580C']
ax.bar(categories, metric_counts, color=colors_cat[:len(categories)])
ax.set_ylabel('Number of Metrics', fontweight='bold')
ax.set_title('Metrics by Category', fontweight='bold')
ax.grid(axis='y', alpha=0.3)
for i, v in enumerate(metric_counts):
    ax.text(i, v + 0.1, str(v), ha='center', fontweight='bold')

# Plot 2: Metric Distribution
ax = axes[0, 1]
category_distribution = [len(metrics_by_category[cat]) for cat in categories]
ax.pie(category_distribution, labels=categories, autopct='%1.1f%%', colors=colors_cat[:len(categories)])
ax.set_title('Distribution of Metrics', fontweight='bold')

# Plot 3: Timeline to First Measurement
ax = axes[1, 0]
frequencies = {}
for metric in success_metrics:
    if metric.frequency not in frequencies:
        frequencies[metric.frequency] = 0
    frequencies[metric.frequency] += 1

freq_labels = list(frequencies.keys())
freq_values = list(frequencies.values())
colors_freq = ['#10B981', '#3B82F6', '#EA580C', '#8B5CF6']
ax.barh(freq_labels, freq_values, color=colors_freq[:len(freq_labels)])
ax.set_xlabel('Number of Metrics', fontweight='bold')
ax.set_title('Measurement Frequency', fontweight='bold')
ax.grid(axis='x', alpha=0.3)

# Plot 4: Metric Categories Summary
ax = axes[1, 1]
ax.axis('off')
summary_text = f"""
SUCCESS METRICS SUMMARY

Total Metrics: {len(success_metrics)}

By Category:
• Quality: {len(metrics_by_category.get('Quality', []))} metrics
• Adoption: {len(metrics_by_category.get('Adoption', []))} metrics
• Functionality: {len(metrics_by_category.get('Functionality', []))} metrics
• Satisfaction: {len(metrics_by_category.get('Satisfaction', []))} metrics
• Business: {len(metrics_by_category.get('Business', []))} metrics

Key Milestones:
✓ Week 1: Launch & track initial adoption
✓ Month 1: Hit activation & completion targets
✓ Month 3: Measure user satisfaction & ROI
✓ Quarter 1: Evaluate business impact
"""
ax.text(0.1, 0.5, summary_text, fontsize=11, family='monospace',
        verticalalignment='center', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.3))

plt.tight_layout()
plt.show()

# Print detailed metrics
print("\n" + "=" * 140)
print("SUCCESS METRICS AND KPIs - DETAILED BREAKDOWN")
print("=" * 140)

for category, metrics in sorted(metrics_by_category.items()):
    print(f"\n📊 {category.upper()} METRICS ({len(metrics)} total)")
    print("-" * 140)
    
    for metric in metrics:
        print(f"\n   {metric.metric_name}")
        print(f"   Target: {metric.target}")
        print(f"   Measurement: {metric.measurement_method}")
        print(f"   Baseline: {metric.baseline}")
        print(f"   Frequency: {metric.frequency} | Owner: {metric.owner}")

print("\n" + "=" * 140)
print("LAUNCH READINESS CHECKLIST - First 30 Days")
print("=" * 140)
print("""
Day 1 (Launch):
  □ Deploy Planner Mode to beta users
  □ Enable analytics tracking
  □ Set up success metrics dashboard
  □ Begin user onboarding

Week 1:
  □ Monitor activation rate (target: 60%+ users create first plan)
  □ Track page load performance (<2 sec)
  □ Monitor critical bug reports
  □ Collect early user feedback

Week 2-4:
  □ Measure plan completion rate (70%+ reach 'Ready to Execute')
  □ Calculate average plan creation time (<15 min)
  □ Run NPS survey
  □ Identify top user issues for Phase 2
  □ Validate AI suggestion accuracy

Post-Launch (Month 2-3):
  □ Quarterly business impact review
  □ Plan feature improvements
  □ Scale to broader user base
  □ Iterate on design/UX based on feedback
""")
````

````python
# ============================================================================
# PART 5: ARCHITECTURE - DESKTOP, MOBILE, CLOUD, SERVER MODELS
# ============================================================================

print("\n\n" + "=" * 130)
print("DISTRIBUTED ARCHITECTURE - DESKTOP | MOBILE | SERVER")
print("=" * 130)

@dataclass
class DeploymentModel:
    """Deployment and sync architecture"""
    model_name: str
    components: str
    data_storage: str
    sync_strategy: str
    use_case: str
    advantages: List[str]
    limitations: List[str]

deployment_models = [
    DeploymentModel(
        model_name="VS Code Extension (Primary)",
        components="Frontend (Vue in WebView) + Local Backend (Node)",
        data_storage="Local filesystem + localStorage",
        sync_strategy="One-way: Export to GitHub/Cloud, Import from Cloud",
        use_case="Individual developer planning within VS Code",
        advantages=[
            "Works offline entirely",
            "No server dependency",
            "Fast local performance",
            "Data stays on machine",
            "Integrated with development flow"
        ],
        limitations=[
            "Single-device only (no cross-device sync)",
            "No real-time collaboration",
            "Manual sync required"
        ]
    ),
    
    DeploymentModel(
        model_name="Desktop App (Electron)",
        components="Full Vue app + Electron container + SQLite local DB",
        data_storage="SQLite local database + file system",
        sync_strategy="Bidirectional: Cloud sync with conflict resolution",
        use_case="Professional users wanting desktop app experience",
        advantages=[
            "Cross-platform (Windows/Mac/Linux)",
            "Local database (structured queries)",
            "Cloud sync capability",
            "Offline-first architecture",
            "Native OS integrations"
        ],
        limitations=[
            "App download/installation required",
            "OS-specific updates needed",
            "Larger footprint"
        ]
    ),
    
    DeploymentModel(
        model_name="Web App (Browser)",
        components="Vue SPA + Laravel backend + PostgreSQL",
        data_storage="PostgreSQL (server) + IndexedDB (browser cache)",
        sync_strategy="Real-time bidirectional sync with WebSockets",
        use_case="Team collaboration, web access from anywhere",
        advantages=[
            "No installation needed",
            "Real-time collaboration",
            "Cross-device seamless sync",
            "Centralized auth & permissions",
            "Easy backups and recovery"
        ],
        limitations=[
            "Requires internet connection",
            "Server uptime dependency",
            "Privacy considerations"
        ]
    ),
    
    DeploymentModel(
        model_name="Mobile App (React Native / Flutter)",
        components="Mobile UI + REST API client",
        data_storage="Local mobile DB (Realm/SQLite) + cloud sync",
        sync_strategy="Background sync, conflict resolution on device",
        use_case="Plan review/approval on mobile, quick edits on-the-go",
        advantages=[
            "Native mobile experience",
            "Push notifications",
            "Gesture-friendly UI",
            "Works offline"
        ],
        limitations=[
            "Limited editing capabilities",
            "Screen size constraints",
            "Focused view vs. detailed planning"
        ]
    )
]

print("\nDEPLOYMENT MODELS:")
print("-" * 130)

for model in deployment_models:
    print(f"\n🖥️  {model.model_name.upper()}")
    print(f"Use Case: {model.use_case}")
    print(f"Components: {model.components}")
    print(f"Storage: {model.data_storage}")
    print(f"Sync: {model.sync_strategy}")
    print(f"\nAdvantages:")
    for adv in model.advantages:
        print(f"  ✓ {adv}")
    print(f"Limitations:")
    for lim in model.limitations:
        print(f"  ⚠ {lim}")

print("\n\n" + "=" * 130)
print("RECOMMENDED DEPLOYMENT STRATEGY")
print("=" * 130)
print("""
PHASE 1 (MVP - March 2026):
✓ VS Code Extension (primary focus)
✓ Web App (secondary, for team access)
- Share plans via export (JSON/Markdown)
- Manual sync via GitHub Issues
- No real-time collaboration

PHASE 2 (Expansion - Q2 2026):
+ Desktop App (Electron wrapper of web app)
+ Cloud sync service (Sync Engine)
- Offline-first strategy
- Conflict resolution for same-plan edits
- Selective sync (sync only your plans)

PHASE 3 (Full Suite - Q3 2026):
+ Mobile App (React Native)
+ Real-time collaboration (WebSockets)
+ Team permissions & roles
+ Activity logging and audit trail

ARCHITECTURE DIAGRAM:
```
┌─────────────────────────────────────────────────────────────────────┐
│                        PLANNER MODE ECOSYSTEM                       │
├─────────────────────────────────────────────────────────────────────┤
│                                                                     │
│  VS Code Extension ──┐                                             │
│  (WebView, Vue)      │                                             │
│                      │  ┌─────────────────────────┐               │
│  Desktop App ────────┼─→│   Sync Engine           │   ┌──────────┐│
│  (Electron, SQLite)  │  │   (Conflict Resolution) │──→│ Cloud DB ││
│                      │  │   (WebSocket Server)    │   │PostgreSQL││
│  Web App ────────────┤  │                         │   └──────────┘│
│  (Browser, Vue)      │  └─────────────────────────┘   ┌──────────┐│
│                      │              ↓                 │ File     ││
│  Mobile App ─────────┘              └────────────────→│ Storage  ││
│  (React Native)                                       │(GitHub)  ││
│                                                       └──────────┘│
└─────────────────────────────────────────────────────────────────────┘
```

DATA FLOW:
1. User creates plan in VS Code
2. Auto-saves to local storage
3. On export, sends to server
4. Server syncs to cloud DB
5. Other devices pull from cloud on login
6. Conflict resolution if same plan edited elsewhere

OFFLINE SUPPORT:
- VS Code: Always works offline
- Desktop: Works offline, syncs when online
- Web: Cached views available offline, edits queued
- Mobile: Background sync when connection restored
""")

# ============================================================================
# PART 6: FRONTEND-BACKEND COMMUNICATION & API CONTRACTS
# ============================================================================

print("\n\n" + "=" * 130)
print("FRONTEND-BACKEND COMMUNICATION ARCHITECTURE")
print("=" * 130)

@dataclass
class APIContract:
    """REST API contract between frontend and backend"""
    endpoint: str
    method: str
    request_body: str
    response_body: str
    error_cases: List[str]

api_contracts = [
    APIContract(
        endpoint="/api/plans",
        method="POST",
        request_body="""
        {
          "name": "Gym Automation System",
          "description": "Build gym membership platform",
          "requirements": "Raw text requirements...",
          "projectType": "feature",
          "teamSize": "medium",
          "dueDate": "2026-02-05"
        }
        """,
        response_body="""
        {
          "id": "plan-001",
          "name": "Gym Automation System",
          "status": "draft",
          "createdAt": "2026-01-08T10:00:00Z",
          "updatedAt": "2026-01-08T10:00:00Z",
          "tasks": [],
          "validationStatus": "pending"
        }
        """,
        error_cases=["401 Unauthorized", "422 Validation Error", "429 Rate Limited"]
    ),
    
    APIContract(
        endpoint="/api/plans/{planId}/tasks",
        method="POST",
        request_body="""
        {
          "title": "Implement OAuth",
          "description": "Add OAuth provider integration",
          "type": "feature",
          "priority": "high",
          "estimatedHours": 8,
          "parentTaskId": "epic-auth",
          "acceptanceCriteria": ["Support Google", "Support GitHub"],
          "dependencies": [{"taskId": "task-auth-setup", "type": "hard"}]
        }
        """,
        response_body="""
        {
          "id": "task-oauth-001",
          "planId": "plan-001",
          "title": "Implement OAuth",
          "type": "feature",
          "priority": "high",
          "estimatedHours": 8,
          "status": "pending",
          "createdAt": "2026-01-08T10:15:00Z"
        }
        """,
        error_cases=["400 Invalid Task", "404 Plan Not Found", "409 Circular Dependency"]
    ),
    
    APIContract(
        endpoint="/api/plans/{planId}/validate",
        method="POST",
        request_body="""
        {
          "validateDependencies": true,
          "validateEstimates": true,
          "validateAcceptanceCriteria": true
        }
        """,
        response_body="""
        {
          "isValid": false,
          "errors": [
            {
              "type": "missing_acceptance_criteria",
              "taskId": "task-payment",
              "message": "Task missing acceptance criteria"
            }
          ],
          "warnings": [
            {
              "type": "long_critical_path",
              "message": "Critical path spans 5 weeks"
            }
          ],
          "score": 75
        }
        """,
        error_cases=["400 Invalid Request", "404 Plan Not Found"]
    ),
    
    APIContract(
        endpoint="/api/plans/{planId}/export",
        method="GET",
        request_body="""
        {
          "format": "json|markdown|csv|pdf",
          "includeMetadata": true
        }
        """,
        response_body="""
        {
          "format": "json",
          "data": "...(full plan JSON)...",
          "filename": "plan-gym-automation.json",
          "generatedAt": "2026-01-08T10:30:00Z"
        }
        """,
        error_cases=["400 Invalid Format", "404 Plan Not Found", "503 Export Service Down"]
    )
]

print("\nKEY API CONTRACTS:")
print("-" * 130)

for contract in api_contracts:
    print(f"\n{contract.method.upper()} {contract.endpoint}")
    print(f"Request: {contract.request_body}")
    print(f"Response: {contract.response_body}")
    print(f"Errors: {', '.join(contract.error_cases)}")

print("\n\n" + "=" * 130)
print("COMMUNICATION PROTOCOL SPECIFICATIONS")
print("=" * 130)
print("""
REST vs. WebSocket Strategy:
┌─────────────────────────────────────────────────────────────────┐
│ OPERATION TYPE        │ PROTOCOL    │ USE CASE                 │
├─────────────────────────────────────────────────────────────────┤
│ Create/Read/Update    │ REST        │ Non-real-time            │
│ List queries          │ REST        │ Initial page load        │
│ File uploads          │ REST        │ Import plans             │
│ Real-time updates     │ WebSocket   │ Live collaboration       │
│ Notifications         │ WebSocket   │ Task comments, approvals │
│ Bulk operations       │ REST Batch  │ Multi-task updates       │
└─────────────────────────────────────────────────────────────────┘

Message Format (JSON):
{
  "id": "msg-123",
  "timestamp": "2026-01-08T10:00:00Z",
  "type": "task_created|task_updated|plan_approved|comment_added",
  "actor": "user-123",
  "resource": {
    "type": "task|plan|comment",
    "id": "task-oauth-001"
  },
  "data": {...},
  "version": 1
}

Error Handling:
{
  "error": {
    "code": "VALIDATION_ERROR",
    "message": "Circular dependency detected",
    "details": {
      "taskChain": ["task-a", "task-b", "task-a"]
    },
    "suggestion": "Remove dependency from task-b to task-a"
  }
}

Rate Limiting:
- 100 requests per minute for authenticated users
- 10 requests per minute for anonymous users
- Batch operations: 1000 items per batch max
- WebSocket: 10 concurrent connections per user

Versioning:
- API v1 (current): All operations
- Backward compatible: Yes
- Deprecation notice: 6 months advance
""")

# Create API summary table
api_data = []
for contract in api_contracts:
    api_data.append({
        'Endpoint': contract.endpoint,
        'Method': contract.method,
        'Purpose': contract.endpoint.split('/')[-1]
    })

df_api = pd.DataFrame(api_data)
print("\n" + "=" * 130)
print("API ENDPOINTS SUMMARY")
print("=" * 130)
print(df_api.to_string(index=False))
````

````python
# ============================================================================
# PART 7: VISUAL REVIEW TOOL - INTERACTIVE PLAN REVIEW WITH ANNOTATIONS
# ============================================================================

print("\n\n" + "=" * 150)
print("🎯 VISUAL REVIEW TOOL - SPECIAL FEATURE FOR PLAN REVIEW & COLLABORATION")
print("=" * 150)

print("""
The Visual Review Tool is a unique feature that enables stakeholders to review plans visually,
make annotations, ask questions, and track changes - all within an interactive visual display.

KEY OBJECTIVE:
Transform plan review from static documents to dynamic, interactive visual feedback loops.

PRIMARY USE CASE:
Product Owner reviewing task decomposition from a designer/planner:
- "This task needs to be smaller"
- "Can we parallel these tasks?"
- "I don't understand this dependency"
- All with visual selection, hover previews, and change tracking
""")

print("\n" + "=" * 150)
print("VISUAL REVIEW TOOL - COMPLETE SPECIFICATION")
print("=" * 150)

@dataclass
class AnnotationType:
    """Types of annotations users can add to visual elements"""
    name: str
    icon: str
    color: str
    use_case: str
    example: str

annotation_types = [
    AnnotationType(
        name="Comment",
        icon="message-circle",
        color="#3B82F6",
        use_case="General feedback or questions about element",
        example="'Can we merge this with task above?'"
    ),
    AnnotationType(
        name="Suggestion",
        icon="lightbulb",
        color="#FBBF24",
        use_case="Propose change to element",
        example="'Reduce from 8 hours to 4 hours'"
    ),
    AnnotationType(
        name="Issue",
        icon="alert-circle",
        color="#DC2626",
        use_case="Flag blocker or problem",
        example="'This creates circular dependency'"
    ),
    AnnotationType(
        name="Approval",
        icon="check-circle",
        color="#10B981",
        use_case="Mark element as approved/reviewed",
        example="'✓ Looks good - approved'"
    ),
    AnnotationType(
        name="Question",
        icon="help-circle",
        color="#8B5CF6",
        use_case="Ask for clarification",
        example='What does "RBAC" mean in this context?'
    ),
    AnnotationType(
        name="Action Item",
        icon="arrow-right-circle",
        color="#06B6D4",
        use_case="Mark as followup action",
        example="'TODO: Get estimation from John before finalizing'"
    )
]

print("\nANNOTATION SYSTEM:")
print("-" * 150)

for annot in annotation_types:
    print(f"\n{annot.icon} {annot.name.upper()}")
    print(f"   Color: {annot.color}")
    print(f"   Use Case: {annot.use_case}")
    print(f"   Example: {annot.example}")

# Annotation capabilities matrix
annotation_features = {
    "Visual Placement": [
        "Appears directly on selected element",
        "Shows author avatar + timestamp",
        "Color-coded by type",
        "Numbered badges for count"
    ],
    "Interaction": [
        "Click element → annotation panel appears",
        "Add new annotation via right-click menu",
        "Reply to existing annotations (threaded)",
        "Resolve annotations when addressed",
        "Filter by annotation type",
        "Filter by author"
    ],
    "Collaboration": [
        "Real-time updates via WebSocket",
        "Multi-user annotation viewing",
        "@mentions to notify team members",
        "History of all annotation activity",
        "Export annotation report to PDF/Markdown"
    ],
    "Integration": [
        "Sync with GitHub Issues (create issue from annotation)",
        "Link to specific plan version",
        "Include in plan diff view",
        "Email notifications for new annotations"
    ]
}

print("\n" + "=" * 150)
print("ANNOTATION FEATURES & CAPABILITIES")
print("=" * 150)

for category, features in annotation_features.items():
    print(f"\n📌 {category.upper()}")
    for feature in features:
        print(f"   ✓ {feature}")

# Mock annotation data
sample_annotations = [
    {
        "id": "ann-001",
        "type": "Question",
        "author": "Alice (Product Owner)",
        "timestamp": "2026-01-08 10:30 AM",
        "target": "Task: Implement User Auth (TASK-001)",
        "content": "Do we need SSO integration from Day 1?",
        "resolved": False,
        "replies": [
            {
                "author": "Bob (Tech Lead)",
                "timestamp": "2026-01-08 10:45 AM",
                "content": "Not for MVP. Can be added in Phase 2."
            }
        ]
    },
    {
        "id": "ann-002",
        "type": "Issue",
        "author": "Charlie (Designer)",
        "timestamp": "2026-01-08 11:15 AM",
        "target": "Task: Design Landing Page (TASK-012)",
        "content": "This task is blocking 3 other tasks but has no assignee!",
        "resolved": True,
        "resolution": "Assigned to Diana (Designer)"
    },
    {
        "id": "ann-003",
        "type": "Suggestion",
        "author": "Eve (Backend Dev)",
        "timestamp": "2026-01-08 02:00 PM",
        "target": "Dependency: TASK-005 → TASK-007",
        "content": "These tasks can be done in parallel if we mock the API.",
        "resolved": False
    }
]

print("\n" + "=" * 150)
print("SAMPLE ANNOTATIONS (MOCK DATA)")
print("=" * 150)

for ann in sample_annotations:
    status = "✅ RESOLVED" if ann.get("resolved") else "🔴 OPEN"
    print(f"\n[{ann['type'].upper()}] {status} - {ann['id']}")
    print(f"Author: {ann['author']} at {ann['timestamp']}")
    print(f"Target: {ann['target']}")
    print(f"Content: {ann['content']}")
    
    if ann.get("replies"):
        print(f"Replies ({len(ann['replies'])}):")
        for reply in ann["replies"]:
            print(f"  → {reply['author']} ({reply['timestamp']}): {reply['content']}")
    
    if ann.get("resolution"):
        print(f"Resolution: {ann['resolution']}")

# Visual layout specification
print("\n" + "=" * 150)
print("VISUAL ANNOTATION UI LAYOUT")
print("=" * 150)

annotation_ui_spec = """
┌────────────────────────────────────────────────────────────────────────────────────┐
│  PLAN VISUAL DISPLAY (Left 70%)         │  ANNOTATION PANEL (Right 30%)           │
├──────────────────────────────────────────┼─────────────────────────────────────────┤
│                                          │  🔍 Filter Annotations:                  │
│  [Network Diagram View]                  │  ☑ Comments  ☑ Questions  ☑ Issues      │
│                                          │  ☐ Suggestions  ☐ Approvals              │
│  • Task nodes with annotation badges    │  ─────────────────────────────────────── │
│  • Hover shows annotation preview        │  📝 Add Annotation:                      │
│  • Click opens full annotation panel     │  [Select Type ▼] [Text Area...]         │
│                                          │  [Submit] [Cancel]                       │
│  Example task with annotations:          │  ─────────────────────────────────────── │
│                                          │  ACTIVE ANNOTATIONS (3):                 │
│  ┌─────────────────────┐                │                                          │
│  │ TASK-001           │ 💬²            │  💬 Question by Alice                    │
│  │ User Auth          │ ❓¹            │  "Do we need SSO integration?"           │
│  │ 8 hours            │                │  [Reply] [Resolve]                       │
│  └─────────────────────┘                │                                          │
│                                          │  ❗ Issue by Charlie                     │
│  Badge indicators:                       │  "No assignee for blocking task"         │
│  💬² = 2 comments                        │  ✅ Resolved                             │
│  ❓¹ = 1 question                        │                                          │
│                                          │  💡 Suggestion by Eve                    │
│                                          │  "Can parallelize with mocking"          │
│                                          │  [Reply] [Resolve]                       │
└──────────────────────────────────────────┴─────────────────────────────────────────┘

INTERACTION FLOW:

1. User hovers over task → Preview tooltip shows annotation count and types
2. User clicks task → Right panel shows all annotations for that element
3. User adds annotation → Choose type → Enter text → Submit
4. Other users see annotation in real-time (WebSocket sync)
5. Annotations can be resolved → Badge count decrements → Moves to "Resolved" section
6. Annotations persist with plan version → Can review historical feedback

ANNOTATION PERSISTENCE:

- Stored in backend database
- Linked to plan version ID
- Includes user identity, timestamp, element target
- Supports threading (replies to annotations)
- Supports resolution workflow
- Exportable to external formats (Markdown, PDF)
"""

print(annotation_ui_spec)

print("\n" + "=" * 150)
print("ANNOTATION WORKFLOW - STEP-BY-STEP")
print("=" * 150)

annotation_workflow = [
    {
        "step": 1,
        "action": "User opens plan in Visual Review mode",
        "system": "Load plan data + all existing annotations from backend",
        "ui": "Display plan with annotation badges on elements"
    },
    {
        "step": 2,
        "action": "User hovers over task",
        "system": "Fetch annotation preview for hovered element",
        "ui": "Show tooltip: '💬 2 comments, ❓ 1 question'"
    },
    {
        "step": 3,
        "action": "User clicks task",
        "system": "Load full annotation details for selected element",
        "ui": "Right panel displays all annotations with threads, timestamps, authors"
    },
    {
        "step": 4,
        "action": "User clicks 'Add Annotation'",
        "system": "Open annotation form",
        "ui": "Form appears: [Type selector] [Text area] [Submit]"
    },
    {
        "step": 5,
        "action": "User submits annotation",
        "system": "POST /api/annotations → Save to DB → Broadcast via WebSocket",
        "ui": "New annotation appears in panel + badge increments + other users notified"
    },
    {
        "step": 6,
        "action": "User resolves annotation",
        "system": "PATCH /api/annotations/{id} → Mark resolved → Update badge count",
        "ui": "Annotation moves to 'Resolved' section + badge decrements"
    },
    {
        "step": 7,
        "action": "User exports annotation report",
        "system": "Generate report with all annotations grouped by task",
        "ui": "Download Markdown or PDF summary of all feedback"
    }
]

print("\nANNOTATION WORKFLOW:")
for flow in annotation_workflow:
    print(f"\nSTEP {flow['step']}: {flow['action']}")
    print(f"   System Action: {flow['system']}")
    print(f"   UI Response: {flow['ui']}")

print("\n" + "=" * 150)
print("TECHNICAL IMPLEMENTATION - ANNOTATION SYSTEM")
print("=" * 150)

annotation_tech_stack = {
    "Backend": [
        "Laravel API endpoints for annotations CRUD",
        "Database schema: annotations table (id, plan_id, element_id, type, author_id, content, resolved, created_at)",
        "WebSocket server for real-time sync (Laravel Echo + Pusher/Soketi)",
        "Notification system for @mentions"
    ],
    "Frontend": [
        "Vue 3 component: AnnotationPanel.vue",
        "Pinia store: annotationStore.ts (state management)",
        "WebSocket client for real-time updates",
        "Rich text editor for annotation content (e.g., TipTap)"
    ],
    "Data Flow": [
        "User adds annotation → POST /api/annotations → DB insert → Broadcast event",
        "Other users subscribed → Receive event → Update local state → Re-render UI",
        "Annotation resolved → PATCH /api/annotations/{id} → Update DB → Broadcast",
        "Export → GET /api/annotations/export → Generate report → Download"
    ]
}

print("\nTECHNICAL ARCHITECTURE:")
for category, details in annotation_tech_stack.items():
    print(f"\n{category.upper()}:")
    for detail in details:
        print(f"   • {detail}")

print("\n" + "=" * 150)
print("END OF VISUAL REVIEW TOOL SPECIFICATION")
print("=" * 150)

print("""
This completes the Visual Review Tool specification.

SUMMARY:
- Annotation system with 6 types (Comment, Suggestion, Issue, Approval, Question, Action Item)
- Visual badges on plan elements showing annotation counts
- Interactive panel for viewing/adding/resolving annotations
- Real-time collaboration via WebSocket
- Threading and resolution workflow
- Export capabilities for reporting
- Integration with GitHub Issues

This feature transforms plan review from static documents to dynamic visual feedback loops,
enabling Product Owners, designers, and stakeholders to collaborate effectively on plan quality.
""")
````

````python
# ============================================================================
# PART 6: DIFFERENT USER JOURNEY PATHS
# ============================================================================

print("\n\n" + "=" * 130)
print("THREE USER JOURNEY TYPES - TAILORED EXPERIENCES")
print("=" * 130)

@dataclass
class UserJourney:
    """Different ways users can go through the design phase"""
    journey_id: str
    user_type: str
    description: str
    primary_focus: List[str]
    skippable_questions: List[int]
    ai_assistance_level: str
    typical_time: str
    required_outputs: List[str]
    strengths: List[str]
    challenges: List[str]
    flow: str

user_journeys = [
    UserJourney(
        journey_id="journey-001",
        user_type="Visual Designer",
        description="I want to design the visual interface first, then worry about logic",
        primary_focus=[
            "Color theme",
            "Page layout",
            "Task display format",
            "Visual hierarchy",
            "Icons & typography"
        ],
        skippable_questions=[7, 8, 9],  # Can skip AI assistance, collaboration, storage
        ai_assistance_level="Light (suggestions only, not required)",
        typical_time="15-20 minutes",
        required_outputs=[
            "Color palette specification",
            "Page mockups",
            "Component library",
            "Interaction specifications"
        ],
        strengths=[
            "✓ Creates beautiful, cohesive interfaces",
            "✓ Fast design iteration",
            "✓ Visual consistency across all pages",
            "✓ Good UX/UI focus"
        ],
        challenges=[
            "✗ May miss technical constraints",
            "✗ Needs technical architect input",
            "✗ Performance considerations overlooked",
            "✗ May need rework in implementation"
        ],
        flow="""
        1. Answer Q1: Page Layout
        2. Answer Q2: Color Theme
        3. Answer Q3: Task Display Format
        4. Answer Q4: Dependency Visualization
        5. Answer Q5: Timeline Representation
        6. [SKIP Q6-9] 
        7. Answer Q10: Your Role = Visual Designer
        8. Design Summary (review visual choices)
        9. AI Asks Follow-Up Questions:
           - Specific color shades?
           - Icon styles?
           - Animation preferences?
           - Responsive breakpoints?
        10. Export: Visual Design Specification
        11. [Next: Share with Technical Architect for Q6-9]
        """
    ),
    
    UserJourney(
        journey_id="journey-002",
        user_type="Business Analyst / Product Owner",
        description="I want to define features and workflows, not worry about visual design",
        primary_focus=[
            "Task decomposition",
            "Dependencies",
            "Timeline & milestones",
            "User workflows",
            "Acceptance criteria"
        ],
        skippable_questions=[2, 3, 4, 5],  # Can skip visual questions
        ai_assistance_level="Heavy (needs all suggestions and AI questions)",
        typical_time="25-35 minutes",
        required_outputs=[
            "Task decomposition",
            "Dependency graph",
            "Timeline/roadmap",
            "Feature requirements",
            "Acceptance criteria"
        ],
        strengths=[
            "✓ Complete requirements definition",
            "✓ Clear task breakdown",
            "✓ Timeline realistic and achievable",
            "✓ Dependencies well-understood"
        ],
        challenges=[
            "✗ May not consider visual constraints",
            "✗ Decisions might affect design",
            "✗ Needs input from Visual Designer",
            "✗ Complex workflows hard to visualize"
        ],
        flow="""
        1. [SKIP Q1-5] (Let designer handle visual)
        2. Answer Q6: User Input Style
        3. Answer Q7: AI Assistance = "Heavy Suggestions"
        4. Answer Q8: Collaboration Model
        5. Answer Q9: Data Storage
        6. Answer Q10: Your Role = Business Analyst
        7. Design Summary (review functional choices)
        8. AI Asks Follow-Up Questions:
           - Task estimation approach?
           - Dependency patterns?
           - Critical path priority?
           - Resource constraints?
           - Integration points?
           - Success metrics?
        9. Export: Requirements & Decomposition
        10. [Next: Share with Designer for Q1-5]
        """
    ),
    
    UserJourney(
        journey_id="journey-003",
        user_type="Hybrid / Technical Architect (All-In-One)",
        description="I want to design everything comprehensively - visual, functional, and technical",
        primary_focus=[
            "All visual aspects",
            "All functional aspects",
            "Technical architecture",
            "Integration patterns",
            "Scalability & performance"
        ],
        skippable_questions=[],  # Answer all questions
        ai_assistance_level="Maximum (deep dive on every topic)",
        typical_time="45-60 minutes",
        required_outputs=[
            "Complete design specification",
            "Visual system",
            "Technical architecture",
            "API contracts",
            "Data models",
            "Deployment strategy"
        ],
        strengths=[
            "✓ Comprehensive, coherent design",
            "✓ No handoff needed",
            "✓ All constraints considered",
            "✓ Implementation-ready specification"
        ],
        challenges=[
            "✗ Very time-consuming",
            "✗ Needs deep expertise across domains",
            "✗ Risk of over-specification",
            "✗ May have analysis paralysis"
        ],
        flow="""
        1. Answer Q1: Page Layout
        2. Answer Q2: Color Theme
        3. Answer Q3: Task Display Format
        4. Answer Q4: Dependency Visualization
        5. Answer Q5: Timeline Representation
        6. Answer Q6: User Input Style
        7. Answer Q7: AI Assistance = "Full Hybrid"
        8. Answer Q8: Collaboration Model
        9. Answer Q9: Data Storage
        10. Answer Q10: Your Role = Holistic/Technical Architect
        11. Design Summary (review all choices)
        12. Deep AI Follow-Up Questions (~10-15):
            - Color contrast accessibility?
            - Mobile responsive strategy?
            - API endpoint design?
            - Database schema?
            - Real-time update mechanisms?
            - Error handling approach?
            - Performance targets?
            - Security considerations?
            - Internationalization?
            - Analytics tracking?
        13. Export: Complete Specification (Visual + Functional + Technical)
        14. Ready for implementation immediately
        """
    )
]

print("\nUSER JOURNEY COMPARISON:")
print("-" * 130)

journey_data = []
for journey in user_journeys:
    journey_data.append({
        'User Type': journey.user_type,
        'Time': journey.typical_time,
        'AI Assistance': journey.ai_assistance_level,
        'Outputs': len(journey.required_outputs),
        'Total Questions': len(journey.skippable_questions)
    })

df_journeys = pd.DataFrame(journey_data)
print(df_journeys.to_string(index=False))

print("\n\n" + "=" * 130)
print("JOURNEY FLOW COMPARISON")
print("=" * 130)

for journey in user_journeys:
    print(f"\n{journey.user_type.upper()}")
    print(f"  Focus: {', '.join(journey.primary_focus[:3])}...")
    print(f"  Time: {journey.typical_time}")
    print(f"  Strengths: {', '.join(j.split('✓ ')[1] for j in journey.strengths[:2])}...")
    print(f"  Outputs: {len(journey.required_outputs)} deliverables")

# ============================================================================
# PART 7: CONTINUOUS DESIGN REFINEMENT CYCLE
# ============================================================================

print("\n\n" + "=" * 130)
print("CONTINUOUS REFINEMENT - ITERATIVE DESIGN PROCESS")
print("=" * 130)

refinement_cycle = """
CONTINUOUS DESIGN LOOP:

Phase 1: Core Questions (10 questions)
├─ User answers or notes initial choices
├─ Visual feedback on each selection
└─ Design summary shows all choices

Phase 2: Design Summary & Review
├─ All 10 core choices displayed side-by-side
├─ Each can be edited (click to jump back to question)
├─ User sees live preview of entire design
├─ User can export at this point if satisfied

Phase 3: AI Contextual Follow-Up (5-15 questions)
├─ AI asks role-specific follow-up questions
├─ Each question based on user's journey type
├─ Side panel auto-shows relevant sections
├─ User can:
│   ├─ Answer with A/B/C/D options
│   ├─ Provide custom/E answer with notes
│   ├─ Skip the question (if optional)
│   ├─ Go back to refine earlier choices
│   └─ Add notes without answering
├─ Real-time visualization updates as answers given
└─ Can loop back to previous questions

Phase 4: Refinement & Iteration
├─ User reviews design with team (if collaboration enabled)
├─ Feedback collected:
│   ├─ Visual feedback (colors, layout)
│   ├─ Functional feedback (workflows, clarity)
│   └─ Technical feedback (feasibility, performance)
├─ Jump back to specific questions
├─ Change answers and see impact
├─ AI suggests alternatives
└─ Iterate until satisfied

Phase 5: Final Design Export
├─ Complete specification document
├─ Visual system (colors, typography, components)
├─ Page mockups (all pages, all themes)
├─ Interaction specifications
├─ Technical architecture (if applicable)
├─ Task decomposition (if applicable)
├─ Dependencies graph
├─ Timeline/roadmap
└─ Export formats: PDF, Markdown, JSON, Figma, etc.

CYCLE CHARACTERISTICS:
✓ Non-linear: Can jump between phases
✓ Iterative: Refine answers and see immediate impact
✓ Collaborative: Easy to share progress and get feedback
✓ Exportable: Export at any point
✓ Resumable: Can save and come back later
✓ Continuous: No hard stop (keep refining until launch)
"""

print(refinement_cycle)

# ============================================================================
# PART 8: FINAL INTERACTIVE DESIGN PHASE SUMMARY
# ============================================================================

print("\n\n" + "=" * 130)
print("INTERACTIVE DESIGN PHASE SUMMARY")
print("=" * 130)

summary = """
═══════════════════════════════════════════════════════════════════════════════════════════════════════════════════════════

WHAT IS THE "SUPER INTERACTIVE DESIGN PHASE"?

The Interactive Design Phase is a guided, question-driven system where users design their planning tool step-by-step with
continuous visual feedback. It's not just Q&A - it's a dynamic experience where every choice immediately affects the visual
preview, allowing users to see (not imagine) how their design decisions work together.

═══════════════════════════════════════════════════════════════════════════════════════════════════════════════════════════

CORE COMPONENTS:

1. STRUCTURED QUESTIONS (10 core + 5-15 contextual)
   • A/B/C/D multiple choice options
   • Custom option (E) for free-form answers
   • Notes section for each question
   • Visual examples and mockups
   • AI follow-up suggestions

2. LIVE PREVIEW SYSTEM
   • Right panel shows affected page sections
   • Real-time updates (200-500ms)
   • Interactive elements (drag, click, hover)
   • Multiple views side-by-side
   • Complete page previews

3. SMART SIDE PANEL
   • Auto-scrolls to affected sections
   • Shows question context
   • Displays AI suggestions
   • Shows design impacts
   • Accumulates user notes

4. AI-POWERED ASSISTANCE
   • Not fully AI-driven (user in control)
   • Contextual suggestions (not overwhelming)
   • Follow-up questions based on answers
   • Compatibility checking
   • Best practice recommendations

5. REAL-TIME FEEDBACK
   • Live color/layout changes
   • Compatibility warnings (⚠️ yellow)
   • Synergy indicators (✓ green)
   • Progress tracking (X of Y questions)
   • Unsaved changes warning

6. FLEXIBLE USER JOURNEYS
   • Visual Designer: 15-20 min, focus on UI/UX
   • Business Analyst: 25-35 min, focus on features
   • Technical Architect: 45-60 min, complete design
   • Can combine approaches (hybrid)

═══════════════════════════════════════════════════════════════════════════════════════════════════════════════════════════

USER EXPERIENCE FLOW:

[START] → [10 Core Questions] → [Design Summary] → [Review & Edit]
                                                          ↓
                                              [Contextual AI Questions]
                                                          ↓
                                              [Refinement & Iteration]
                                                          ↓
                                              [Final Export/Share]

Each step:
✓ Shows live preview of design
✓ Allows jumping back to previous questions
✓ Displays side panel with context
✓ Updates design across all panels
✓ Accumulates user notes
✓ Checks for conflicts/synergies
✓ Suggests improvements

═══════════════════════════════════════════════════════════════════════════════════════════════════════════════════════════

KEY FEATURES THAT MAKE IT "SUPER INTERACTIVE":

1. ONE QUESTION AT A TIME
   • Focused, not overwhelming
   • Clear, actionable options
   • Room for custom answers

2. CONTINUOUS VISUALIZATION
   • See design changes instantly
   • All pages update together
   • Interactive elements respond immediately
   • Different view types (preview, interactive, demo)

3. SMART NAVIGATION
   • Side panel shows relevant section
   • Auto-scrolls to affected area
   • Highlights related UI elements
   • Can interact with preview (test before finalizing)

4. COMPREHENSIVE NOTES
   • Every question has notes section
   • Accumulate in side panel
   • Reference throughout design
   • Export with specification

5. FLEXIBLE DEPTH
   • Skip non-critical questions
   • Dive deep on what matters
   • Different paths for different users
   • Export incomplete designs

═══════════════════════════════════════════════════════════════════════════════════════════════════════════════════════════

INTERACTION PATTERNS:

Click Option → Real-time Preview Update → Check Side Panel → Add Note → Next Question

                                              ↓ (or jump back)

Can Edit Previous Answer → Preview Updates → Check Impacts → Continue

═══════════════════════════════════════════════════════════════════════════════════════════════════════════════════════════

OUTPUTS AT EACH STAGE:

After Core Questions: Design Summary + Concept Document
After AI Follow-Ups: Detailed Specification + Visual System
After Refinement: Production-Ready Design + Implementation Guide
Final Export: 
   • PDF Design Document
   • JSON Configuration
   • Figma Design File
   • Markdown Specification
   • Task Decomposition (if applicable)
   • Team-Shareable Document

═══════════════════════════════════════════════════════════════════════════════════════════════════════════════════════════

WHY THIS APPROACH?

Traditional design process: Standalone tool → Mockups → Discussion → Revision → Repeat
This approach: Guided questions → Instant visualization → Continuous feedback → Immediate export

Benefits:
✓ Faster: See impact instantly, not in separate mockup
✓ Clearer: Visual proof, not imagined concepts
✓ More inclusive: Non-designers can participate (just answer questions)
✓ Flexible: Different paths for different expertise levels
✓ Complete: Covers all aspects (visual, functional, technical)
✓ Exportable: Production-ready specs from day one
✓ Collaborative: Easy to share and get feedback
✓ Iterative: Keep refining until perfect

═══════════════════════════════════════════════════════════════════════════════════════════════════════════════════════════
"""

print(summary)
````

## Section 9.1: Interactive Design Phase - Technical Architecture

### Overview of the "Super Interactive" Experience

The Interactive Design Phase is the **core differentiator** of Planner Mode. It transforms planning from a tedious form-filling experience into an engaging, visual, guided journey where every decision is immediately visualized and validated.

### Real-Time Update Cascade

When user selects an option:

1. **LEFT PANEL**: Selection highlighted (instant)
2. **CENTER PANEL**: Preview updates (200-500ms animation)
   - Colors change
   - Layout shifts
   - Elements appear/disappear
   - Interactive demo shows the choice in action
3. **RIGHT PANEL**: Auto-scrolls to relevant section (immediate)
   - Shows which section this affects
   - Highlights elements on preview
   - AI generates contextual questions
   - Notes accumulate
4. **FEEDBACK**: 
   - Compatibility badge (✓ good or ⚠️ warning)
   - Progress bar updates
   - Unsaved indicator appears

Total perceived delay: **0ms** (feels instant)

As the user goes through questions, the right panel **automatically shows and highlights** the relevant section:

**Question 3** → Right Panel shows: "Task List Display" section with different formats
**Question 4** → Right Panel shows: "Dependency Graph" section with visualization styles

When user adds a note or AI asks a follow-up, the panel **auto-scrolls** to relevant parts, so they can see:
- What was asked

---

### Real-Time Feedback Mechanisms

#### Visual Feedback
- **Color Change**: Theme selection instantly recolors preview
- **Layout Shift**: Display format changes trigger layout reorganization
- **Element Appear**: New UI elements fade in
- **Animation**: Staggered entrance for visual hierarchy

#### Compatibility Feedback
- **✓ Green checkmark**: Synergistic choices (e.g., Sidebar + Light theme)
- **⚠️ Yellow warning**: Conflicting choices (e.g., Kanban + Network graph)
- **Suggested alternative**: AI recommends better combination
- **Explanation**: Why this combination works/doesn't work

#### Progress Feedback
- **Progress bar**: Visual indicator of completion (5/10 questions)
- **Time estimate**: "~10-15 minutes remaining"
- **Checkmarks**: Completed questions marked ✓
- **Unsaved indicator**: Dot showing unsaved changes

#### State Feedback
- **Toast notifications**: Important state changes
- **Inline alerts**: Conflicts or warnings
- **Status badges**: Question has notes (orange dot)
- **Disabled options**: Grayed out if conflicting with previous choices

---

### Key UX Principles

1. **One Question at a Time**
   - Not overwhelming
   - Clear focus
   - Room for detailed notes

2. **Continuous Visualization**
   - Not "imagine how this will look"
   - But "here's exactly how it will look"
   - Interactive testing before finalizing

3. **Guided But Not Forced**
   - Can skip non-critical questions
   - Can answer with custom option
   - Can go back and change any answer

4. **Clear Consequences**
   - See impact of each choice
   - Understand compatibility
   - Learn best practices from AI

5. **Progressive Disclosure**
   - Core questions first (10)
   - Then contextual questions (5-15)
   - Then refinement loop
   - Export when ready

---

### Integration Points

#### With Visual Review Tool
After designing, users can:
- Export design to Visual Review Tool
- Have team annotate and comment
- Refine based on feedback
- Jump back to planner to edit

#### With Task Decomposition
If designing a task planning tool:
- Design visual system first
- Then decompose tasks using designed interface
- Tasks show in designed layout/colors
- Consistency guaranteed

#### With Implementation
- Export specs to developers
- Developers have complete visual + functional spec
- No guesswork about design intent
- Faster, more accurate implementation

---

### Technical Implementation Notes

- **Left Panel**: Vue component for question rendering + markdown notes
- **Center Panel**: Real-time preview using actual Vue components (not mockups)
- **Right Panel**: Smart scrolling container with section highlights
- **Update Cascade**: Event-driven, debounced (100-500ms) to prevent flicker
- **State Management**: Pinia store tracks all questions + answers + notes
- **Export**: Generate PDF/JSON/Markdown from final state
- **Persistence**: Auto-save to localStorage (draft) + cloud (if enabled)

---

### Success Criteria for Interactive Design Phase

✓ User completes core questions in < 20 minutes
✓ User understands their design choices immediately (visual proof)
✓ User can export production-ready specification
✓ User can involve team and refine based on feedback
✓ Visual Designer path: Can design without developer input
✓ Analyst path: Can define requirements without designer input
✓ Architect path: Complete spec with all aspects covered
✓ No confusion about design intent or interaction model
✓ Developers can implement directly from exported spec
✓ Design consistency maintained across all pages/themes

```
│  │     - design-system.json (visual assets)                     │          │
│  │     - metadata.json (version, author, timestamps)            │          │
│  └──────────────────────────────────────────────────────────────┘          │
│                           ↓                                                 │
│  Phase 3: DISCOVERY                                                         │
│  ┌──────────────────────────────────────────────────────────────┐          │
│  │ On workspace open, scan for plans:                           │          │
│  │   1. Check /Docs/Plans/ (primary)                            │          │
│  │   2. Check root folder for .plans/ or Plans/ folders         │          │
│  │   3. Check for plan.json files anywhere in tree              │          │
│  │   4. Parse and index all found plans                         │          │
│  └──────────────────────────────────────────────────────────────┘          │
│                           ↓                                                 │
│  Phase 4: LOADING                                                           │
│  ┌──────────────────────────────────────────────────────────────┐          │
│  │ Parse plan files:                                            │          │
│  │   - Extract design choices                                   │          │
│  │   - Load task decomposition (if exists)                      │          │
│  │   - Identify completed vs. pending work                      │          │
│  │   - Check for version conflicts                              │          │
│  │   - Validate plan integrity                                  │          │
│  └──────────────────────────────────────────────────────────────┘          │
│                           ↓                                                 │
│  Phase 5: EXECUTION                                                         │
│  ┌──────────────────────────────────────────────────────────────┐          │
│  │ Plan drives orchestration:                                   │          │
│  │   → Task decomposition (based on plan)                       │          │
│  │   → Dependency mapping                                       │          │
│  │   → Auto Zen gets task list                                  │          │
│  │   → Code implementation guided by specs                      │          │
│  │   → Testing validates against acceptance criteria            │          │
│  └──────────────────────────────────────────────────────────────┘          │
│                           ↓                                                 │
│  Phase 6: EVOLUTION                                                         │
│  ┌──────────────────────────────────────────────────────────────┐          │
│  │ Plan changes over time:                                      │          │
│  │   - User updates design choices                              │          │
│  │   - New requirements discovered                              │          │
│  │   - Plan version incremented                                 │          │
│  │   - Code sync process triggered                              │          │
│  │   - Tasks updated to match new plan                          │          │
│  └──────────────────────────────────────────────────────────────┘          │
│                           ↓                                                 │
│  Phase 7: SYNCHRONIZATION                                                   │
│  ┌──────────────────────────────────────────────────────────────┐          │
│  │ Keep code aligned with plan:                                │          │
│  │   - Detect plan changes (diff from last version)             │          │
│  │   - Analyze impact on existing code                          │          │
│  │   - Generate tasks for code updates                          │          │
│  │   - Auto Zen implements changes                              │          │
│  │   - Validate implementation matches plan                     │          │
│  └──────────────────────────────────────────────────────────────┘          │
│                           ↓                                                 │
│  Phase 8: COMPLETION                                                        │
│  ┌──────────────────────────────────────────────────────────────┐          │
│  │ All tasks completed:                                         │          │
│  │   - Mark plan as "implemented"                               │          │
│  │   - Archive plan version                                     │          │
│  │   - Generate completion report                               │          │
│  │   - Optional: Create new plan for next iteration             │          │
│  └──────────────────────────────────────────────────────────────┘          │
│                                                                             │
└─────────────────────────────────────────────────────────────────────────────┘
```

---

## Why This Matters

Traditional approach:
- Design → Implement → Requirements change → Code gets out of sync → Chaos

**Planner Mode approach:**
- Design → Implement → Requirements change → **Update plan** → **Code auto-updates** → Alignment maintained

The plan is the **single source of truth** that drives everything:
- Task decomposition
- AI agent direction
- Testing validation
- Issue detection
- Code synchronization

When the plan changes, everything else adapts automatically.

````python
# ============================================================================
# PART 1: PLAN DISCOVERY & LOADING
# ============================================================================

from dataclasses import dataclass
from typing import List, Dict, Optional
from datetime import datetime
import pandas as pd

@dataclass
class PlanLocation:
    """Where to look for plan files"""
    location_id: str
    path_pattern: str
    priority: int  # 1 = highest priority
    description: str
    file_patterns: List[str]
    required_files: List[str]
    validation_rules: List[str]

plan_locations = [
    PlanLocation(
        location_id="loc-001",
        path_pattern="Docs/Plans/**",
        priority=1,
        description="Primary plan storage location (recommended)",
        file_patterns=["plan.json", "plan.md", "*.plan.json"],
        required_files=["plan.json"],
        validation_rules=[
            "Must have plan.json",
            "plan.json must have version field",
            "plan.json must have planName field"
        ]
    ),
    
    PlanLocation(
        location_id="loc-002",
        path_pattern="Docs/plans/**",  # lowercase variant
        priority=2,
        description="Alternative location (lowercase)",
        file_patterns=["plan.json", "plan.md", "*.plan.json"],
        required_files=["plan.json"],
        validation_rules=[
            "Must have plan.json",
            "plan.json must have version field"
        ]
    ),
    
    PlanLocation(
        location_id="loc-003",
        path_pattern=".plans/**",
        priority=3,
        description="Hidden folder in root (dot-prefix)",
        file_patterns=["plan.json", "*.plan.json"],
        required_files=["plan.json"],
        validation_rules=[
            "Must have plan.json"
        ]
    ),
    
    PlanLocation(
        location_id="loc-004",
        path_pattern="Plans/**",
        priority=4,
        description="Root-level Plans folder",
        file_patterns=["plan.json", "*.plan.json"],
        required_files=["plan.json"],
        validation_rules=[
            "Must have plan.json"
        ]
    ),
    
    PlanLocation(
        location_id="loc-005",
        path_pattern="**/*.plan.json",
        priority=5,
        description="Anywhere in workspace (fallback scan)",
        file_patterns=["*.plan.json"],
        required_files=["*.plan.json"],
        validation_rules=[
            "File must end with .plan.json",
            "Must be valid JSON"
        ]
    )
]

print("=" * 130)
print("PLAN DISCOVERY - WHERE TO LOOK FOR EXISTING PLANS")
print("=" * 130)

loc_data = []
for loc in plan_locations:
    loc_data.append({
        'Priority': loc.priority,
        'Path Pattern': loc.path_pattern,
        'Description': loc.description,
        'Required Files': ', '.join(loc.required_files)
    })

df_locations = pd.DataFrame(loc_data)
print("\nPLAN SEARCH LOCATIONS (in priority order):")
print("-" * 130)
print(df_locations.to_string(index=False))

# ============================================================================
# PART 2: PLAN FILE STRUCTURE
# ============================================================================

print("\n\n" + "=" * 130)
print("PLAN FILE STRUCTURE - WHAT GETS SAVED")
print("=" * 130)

plan_structure = {
    "directory": "Docs/Plans/{plan-name}/",
    "files": {
        "plan.json": {
            "purpose": "Machine-readable plan configuration",
            "format": "JSON",
            "required": True,
            "schema": {
                "planId": "string (UUID)",
                "planName": "string",
                "version": "string (semver: 1.0.0)",
                "createdAt": "ISO8601 timestamp",
                "updatedAt": "ISO8601 timestamp",
                "author": "string",
                "status": "draft | active | implemented | archived",
                "designChoices": {
                    "pageLayout": "sidebar | tabs | wizard | custom",
                    "colorTheme": "light | dark | highContrast | custom",
                    "taskDisplay": "tree | kanban | grid | custom",
                    "dependencyVisualization": "network | hierarchical | timeline | list",
                    "timelineRepresentation": "gantt | linear | kanban | calendar",
                    "userInputStyle": "inline | modal | sidebar | fullPage",
                    "aiAssistanceLevel": "manual | suggestions | smartDefaults | hybrid",
                    "collaborationModel": "solo | teamAsync | realtime | custom",
                    "dataStorage": "local | cloud | hybrid | customBackend"
                },
                "customConfiguration": {},
                "notes": {},
                "taskDecomposition": [],
                "dependencies": [],
                "timeline": {},
                "metadata": {}
            }
        },
        
        "plan.md": {
            "purpose": "Human-readable plan specification",
            "format": "Markdown",
            "required": False,
            "sections": [
                "# Plan Overview",
                "## Design Choices",
                "## Visual System",
                "## Page Specifications",
                "## Task Decomposition",
                "## Dependencies",
                "## Timeline",
                "## Technical Notes",
                "## Implementation Guide"
            ]
        },
        
        "design-system.json": {
            "purpose": "Visual design assets and specifications",
            "format": "JSON",
            "required": False,
            "schema": {
                "colors": {
                    "primary": "#3B82F6",
                    "secondary": "#8B5CF6",
                    "accent": "#10B981",
                    "backgrounds": {},
                    "text": {},
                    "borders": {},
                    "status": {}
                },
                "typography": {
                    "fontFamily": "Inter, system-ui",
                    "headingSizes": {},
                    "bodySizes": {},
                    "lineHeights": {}
                },
                "spacing": {},
                "components": {},
                "icons": {}
            }
        },
        
        "metadata.json": {
            "purpose": "Version tracking and change history",
            "format": "JSON",
            "required": True,
            "schema": {
                "currentVersion": "1.0.0",
                "versionHistory": [
                    {
                        "version": "1.0.0",
                        "timestamp": "ISO8601",
                        "author": "string",
                        "changes": ["list of changes"],
                        "codeSync": {
                            "required": True,
                            "completed": False,
                            "tasksGenerated": []
                        }
                    }
                ],
                "implementationStatus": {
                    "totalTasks": 0,
                    "completedTasks": 0,
                    "inProgressTasks": 0,
                    "blockedTasks": 0
                },
                "lastSyncedAt": "ISO8601 timestamp or null"
            }
        },
        
        "tasks.json": {
            "purpose": "Generated task decomposition from plan",
            "format": "JSON",
            "required": False,
            "schema": {
                "tasks": [
                    {
                        "taskId": "string",
                        "title": "string",
                        "description": "string",
                        "status": "pending | in-progress | done | blocked",
                        "priority": "critical | high | medium | low",
                        "dependencies": ["taskId1", "taskId2"],
                        "acceptanceCriteria": [],
                        "generatedFromPlan": True,
                        "planVersion": "1.0.0"
                    }
                ]
            }
        }
    }
}

print(f"\nPLAN DIRECTORY: {plan_structure['directory']}")
print("\nFILE BREAKDOWN:")
print("-" * 130)

for filename, details in plan_structure['files'].items():
    required_badge = "✓ REQUIRED" if details['required'] else "○ Optional"
    print(f"\n{filename} ({details['format']}) - {required_badge}")
    print(f"  Purpose: {details['purpose']}")
    if 'sections' in details:
        print(f"  Sections: {len(details['sections'])}")
    if 'schema' in details:
        print(f"  Schema: {len(details['schema'])} top-level fields")

# ============================================================================
# PART 3: PLAN DISCOVERY ALGORITHM
# ============================================================================

print("\n\n" + "=" * 130)
print("PLAN DISCOVERY ALGORITHM")
print("=" * 130)

discovery_algorithm = """
PLAN DISCOVERY PROCESS:

1. ON WORKSPACE OPEN:
   ├─ Trigger: VS Code workspace opened
   ├─ Action: Start plan discovery scan
   └─ Timeout: 5 seconds max (don't block startup)

2. SCAN LOCATIONS (in priority order):
   ├─ Priority 1: Docs/Plans/** (recommended)
   │   ├─ Look for folders containing plan.json
   │   ├─ Validate structure
   │   └─ Add to plan index
   │
   ├─ Priority 2: Docs/plans/** (lowercase variant)
   │   └─ Same as Priority 1
   │
   ├─ Priority 3: .plans/** (hidden folder)
   │   └─ Same as Priority 1
   │
   ├─ Priority 4: Plans/** (root level)
   │   └─ Same as Priority 1
   │
   └─ Priority 5: **/*.plan.json (anywhere)
       ├─ Fallback: scan entire workspace
       ├─ Find any file ending in .plan.json
       └─ Parse and validate

3. FOR EACH FOUND PLAN:
   ├─ Read plan.json
   ├─ Validate schema
   ├─ Check version format (semver)
   ├─ Load metadata.json
   ├─ Check implementation status
   ├─ Parse task decomposition (if exists)
   ├─ Add to plan index
   └─ Show in UI (Plans panel)

4. PLAN INDEX:
   ├─ Store in memory (fast access)
   ├─ Cache to disk (persist between sessions)
   ├─ Update on file changes (watch mode)
   └─ Refresh on user request

5. USER NOTIFICATIONS:
   ├─ If 0 plans found: "No plans found. Create your first plan?"
   ├─ If 1 plan found: "Loaded plan: {plan-name}"
   ├─ If 2+ plans found: "Found {count} plans. Select one to activate."
   └─ If plan invalid: "Plan {name} has validation errors. Review?"

6. ERROR HANDLING:
   ├─ Invalid JSON: Show error, offer to fix
   ├─ Missing required fields: Show warning, provide defaults
   ├─ Version mismatch: Offer migration
   └─ Corrupted file: Offer restore from backup
"""

print(discovery_algorithm)

# ============================================================================
# PART 4: PLAN LOADING & PARSING
# ============================================================================

print("\n\n" + "=" * 130)
print("PLAN LOADING & PARSING - UNDERSTANDING WHAT'S BEEN DONE")
print("=" * 130)

@dataclass
class PlanLoadingStep:
    """Steps to load and understand an existing plan"""
    step_id: str
    step_number: int
    name: str
    description: str
    input_files: List[str]
    actions: List[str]
    output: str
    error_handling: str

loading_steps = [
    PlanLoadingStep(
        step_id="load-001",
        step_number=1,
        name="Read plan.json",
        description="Load the core plan configuration",
        input_files=["plan.json"],
        actions=[
            "Read file from disk",
            "Parse JSON",
            "Validate schema",
            "Extract planId, version, status"
        ],
        output="PlanConfig object in memory",
        error_handling="If invalid JSON → show error, offer to recreate"
    ),
    
    PlanLoadingStep(
        step_id="load-002",
        step_number=2,
        name="Load metadata.json",
        description="Get version history and implementation status",
        input_files=["metadata.json"],
        actions=[
            "Read metadata file",
            "Parse version history",
            "Get current version",
            "Check implementation status",
            "Identify last sync timestamp"
        ],
        output="Metadata object with version history",
        error_handling="If missing → create default metadata from plan.json"
    ),
    
    PlanLoadingStep(
        step_id="load-003",
        step_number=3,
        name="Parse design choices",
        description="Understand what the user designed",
        input_files=["plan.json"],
        actions=[
            "Extract designChoices section",
            "Map to design questions answered",
            "Identify custom configurations",
            "Load user notes from design phase"
        ],
        output="DesignChoices object",
        error_handling="If incomplete → mark as partial design"
    ),
    
    PlanLoadingStep(
        step_id="load-004",
        step_number=4,
        name="Load task decomposition",
        description="Get the list of tasks generated from plan",
        input_files=["tasks.json", "plan.json"],
        actions=[
            "Read tasks.json if exists",
            "Parse task list",
            "Check task statuses",
            "Identify completed vs pending",
            "Map dependencies"
        ],
        output="TaskList with status for each task",
        error_handling="If tasks.json missing → offer to generate from plan"
    ),
    
    PlanLoadingStep(
        step_id="load-005",
        step_number=5,
        name="Detect implementation progress",
        description="Figure out what's been built already",
        input_files=["metadata.json", "tasks.json", "codebase files"],
        actions=[
            "Count total tasks",
            "Count completed tasks",
            "Identify in-progress work",
            "Check for code matching plan specs",
            "Estimate completion percentage"
        ],
        output="ImplementationStatus object",
        error_handling="If ambiguous → ask user to clarify status"
    ),
    
    PlanLoadingStep(
        step_id="load-006",
        step_number=6,
        name="Load visual design system",
        description="Get colors, typography, components",
        input_files=["design-system.json", "plan.json"],
        actions=[
            "Read design-system.json if exists",
            "Parse color palette",
            "Load typography settings",
            "Extract component specifications",
            "Map to design choices"
        ],
        output="DesignSystem object",
        error_handling="If missing → generate from designChoices in plan.json"
    ),
    
    PlanLoadingStep(
        step_id="load-007",
        step_number=7,
        name="Validate plan integrity",
        description="Ensure plan is complete and consistent",
        input_files=["All plan files"],
        actions=[
            "Check all required fields present",
            "Validate version consistency",
            "Check for circular dependencies",
            "Verify task references",
            "Validate design choice compatibility"
        ],
        output="ValidationReport (errors/warnings)",
        error_handling="If errors → show to user, offer auto-fix"
    ),
    
    PlanLoadingStep(
        step_id="load-008",
        step_number=8,
        name="Index and activate plan",
        description="Make plan available in UI",
        input_files=["All loaded data"],
        actions=[
            "Add to plan index",
            "Set as active plan (if only one)",
            "Show in Plans panel",
            "Enable plan-related features",
            "Start file watchers for changes"
        ],
        output="Plan available in UI, ready for use",
        error_handling="If conflicts with other plans → ask user which to activate"
    )
]

print("\nPLAN LOADING STEPS:")
print("-" * 130)

for step in loading_steps:
    print(f"\n[STEP {step.step_number}] {step.name}")
    print(f"  Description: {step.description}")
    print(f"  Input: {', '.join(step.input_files)}")
    print(f"  Output: {step.output}")
    print(f"  Error Handling: {step.error_handling}")

print("\n\n" + "=" * 130)
print("LOADING FLOW DIAGRAM")
print("=" * 130)

loading_flow = """
Plan Discovered
    ↓
Read plan.json
    ├─ Valid? → Continue
    └─ Invalid? → Show error, offer fix
    ↓
Load metadata.json
    ├─ Exists? → Parse version history
    └─ Missing? → Create default metadata
    ↓
Parse design choices
    ├─ Extract what user designed
    ├─ Identify custom configurations
    └─ Load notes
    ↓
Load task decomposition
    ├─ tasks.json exists? → Parse tasks
    └─ Missing? → Offer to generate from plan
    ↓
Detect implementation progress
    ├─ Count completed tasks
    ├─ Identify in-progress work
    └─ Estimate completion %
    ↓
Load design system
    ├─ design-system.json exists? → Parse visuals
    └─ Missing? → Generate from design choices
    ↓
Validate integrity
    ├─ All valid? → Continue
    └─ Errors? → Show report, offer fixes
    ↓
Index and activate
    ├─ Add to plan index
    ├─ Show in UI
    ├─ Enable features
    └─ Start watching for changes
    ↓
PLAN READY TO USE
    ├─ Can view design
    ├─ Can see tasks
    ├─ Can track progress
    ├─ Can make updates
    └─ Can sync with code
"""

print(loading_flow)
````

````python
# ============================================================================
# PART 5: PLAN VERSIONING & CHANGE TRACKING
# ============================================================================

print("\n\n" + "=" * 130)
print("PLAN VERSIONING - TRACKING CHANGES OVER TIME")
print("=" * 130)

@dataclass
class PlanVersion:
    """A version in the plan's history"""
    version: str  # semver: 1.0.0, 1.1.0, 2.0.0
    timestamp: str
    author: str
    changes: List[str]
    change_type: str  # "major" | "minor" | "patch"
    requires_code_sync: bool
    sync_completed: bool
    generated_tasks: List[str]

version_examples = [
    PlanVersion(
        version="1.0.0",
        timestamp="2026-01-08T14:30:00Z",
        author="User",
        changes=["Initial plan creation", "Completed Interactive Design Phase"],
        change_type="major",
        requires_code_sync=True,
        sync_completed=False,
        generated_tasks=[]
    ),
    
    PlanVersion(
        version="1.1.0",
        timestamp="2026-01-10T09:15:00Z",
        author="User",
        changes=[
            "Changed color theme from Light to Dark",
            "Updated primary color to purple",
            "Added custom sidebar collapse behavior"
        ],
        change_type="minor",
        requires_code_sync=True,
        sync_completed=False,
        generated_tasks=["TASK-001: Update color palette", "TASK-002: Implement sidebar collapse"]
    ),
    
    PlanVersion(
        version="1.1.1",
        timestamp="2026-01-10T11:45:00Z",
        author="User",
        changes=[
            "Fixed typo in task description",
            "Updated acceptance criteria wording"
        ],
        change_type="patch",
        requires_code_sync=False,
        sync_completed=True,
        generated_tasks=[]
    ),
    
    PlanVersion(
        version="2.0.0",
        timestamp="2026-01-15T16:00:00Z",
        author="User",
        changes=[
            "Complete redesign: Changed from Kanban to Tree view",
            "Switched from Sidebar to Tabs navigation",
            "Added real-time collaboration features",
            "Changed storage from Local to Cloud"
        ],
        change_type="major",
        requires_code_sync=True,
        sync_completed=False,
        generated_tasks=[
            "TASK-010: Refactor task display to tree view",
            "TASK-011: Replace sidebar with tabs",
            "TASK-012: Implement WebSocket for real-time",
            "TASK-013: Add cloud storage backend"
        ]
    )
]

print("\nVERSION HISTORY EXAMPLE:")
print("-" * 130)

for v in version_examples:
    sync_status = "✓ Synced" if v.sync_completed else "⚠ Pending sync"
    print(f"\nVersion {v.version} ({v.change_type}) - {sync_status}")
    print(f"  Date: {v.timestamp}")
    print(f"  Changes: {len(v.changes)}")
    for change in v.changes:
        print(f"    - {change}")
    if v.generated_tasks:
        print(f"  Generated Tasks: {len(v.generated_tasks)}")
        for task in v.generated_tasks[:2]:
            print(f"    → {task}")

# ============================================================================
# VERSIONING RULES
# ============================================================================

print("\n\n" + "=" * 130)
print("VERSIONING RULES (Semantic Versioning)")
print("=" * 130)

versioning_rules = """
SEMANTIC VERSIONING: MAJOR.MINOR.PATCH

MAJOR VERSION (x.0.0):
├─ Triggered by:
│   ├─ Complete redesign (changing multiple core design choices)
│   ├─ Changing architecture fundamentally
│   ├─ Breaking changes to existing implementation
│   └─ Major feature additions that change the product
├─ Examples:
│   ├─ Switching from Kanban to Tree view
│   ├─ Changing from Solo to Real-time Collaboration
│   ├─ Adding multi-tenant support
│   └─ Complete UI overhaul
├─ Code Sync Impact: HIGH
│   ├─ Major refactoring required
│   ├─ Many tasks generated
│   └─ Extensive testing needed
└─ User Action: Requires explicit confirmation

MINOR VERSION (1.x.0):
├─ Triggered by:
│   ├─ New features added
│   ├─ Design choice changes (non-breaking)
│   ├─ Enhancements to existing functionality
│   └─ Configuration updates
├─ Examples:
│   ├─ Changing color theme
│   ├─ Adding new page/section
│   ├─ Updating task display options
│   └─ Enabling AI suggestions
├─ Code Sync Impact: MEDIUM
│   ├─ Some refactoring needed
│   ├─ Few tasks generated
│   └─ Focused testing
└─ User Action: Auto-increment with user notification

PATCH VERSION (1.1.x):
├─ Triggered by:
│   ├─ Bug fixes in plan
│   ├─ Typo corrections
│   ├─ Documentation updates
│   ├─ Clarifications (no functional change)
│   └─ Metadata updates
├─ Examples:
│   ├─ Fixing typos
│   ├─ Updating task descriptions
│   ├─ Clarifying acceptance criteria
│   └─ Adding notes/comments
├─ Code Sync Impact: LOW or NONE
│   ├─ No code changes needed
│   ├─ Documentation updates only
│   └─ No testing required
└─ User Action: Auto-increment (silent)

VERSION INCREMENT LOGIC:

If user changes:
  ├─ Page layout (sidebar → tabs) → MAJOR
  ├─ Task display (tree → kanban) → MAJOR (if code exists) or MINOR (if not implemented)
  ├─ Color theme → MINOR
  ├─ Add AI assistance → MINOR
  ├─ Fix typo in task → PATCH
  └─ Update notes → PATCH

Auto-increment rules:
  ├─ Detect which fields changed
  ├─ Calculate impact score
  ├─ Determine version type
  ├─ Increment version
  ├─ Generate changelog
  └─ Notify user
"""

print(versioning_rules)

# ============================================================================
# PART 6: PLAN UPDATES & CODE SYNCHRONIZATION
# ============================================================================

print("\n\n" + "=" * 130)
print("PLAN UPDATES & CODE SYNCHRONIZATION")
print("=" * 130)

@dataclass
class PlanUpdateFlow:
    """What happens when user updates a plan"""
    flow_id: str
    trigger: str
    detection: str
    analysis: List[str]
    sync_actions: List[str]
    task_generation: str
    completion: str

update_flows = [
    PlanUpdateFlow(
        flow_id="update-001",
        trigger="User changes color theme from Light to Dark",
        detection="File watcher detects plan.json modified",
        analysis=[
            "Compare new plan.json with previous version",
            "Identify changed field: designChoices.colorTheme",
            "Calculate impact: Affects all pages, all components",
            "Determine version increment: MINOR (1.0.0 → 1.1.0)",
            "Check if code exists for this plan"
        ],
        sync_actions=[
            "Generate task: Update global color palette",
            "Generate task: Update all component theme variables",
            "Generate task: Test dark theme accessibility",
            "Add tasks to _ZENTASKS/tasks.json",
            "Mark tasks as generated from plan v1.1.0"
        ],
        task_generation="3 tasks created",
        completion="Auto Zen picks up tasks and implements"
    ),
    
    PlanUpdateFlow(
        flow_id="update-002",
        trigger="User switches from Kanban to Tree view for tasks",
        detection="File watcher detects plan.json modified",
        analysis=[
            "Compare versions",
            "Changed field: designChoices.taskDisplay",
            "Impact: Major refactor of task display component",
            "Version increment: MAJOR (1.1.0 → 2.0.0)",
            "Existing code affected: TaskList.vue, TaskCard.vue"
        ],
        sync_actions=[
            "Generate task: Refactor TaskList from Kanban to Tree",
            "Generate task: Update TaskCard for tree display",
            "Generate task: Add expand/collapse functionality",
            "Generate task: Update tests for tree view",
            "Generate task: Migrate user data (if needed)",
            "Create backup of current implementation",
            "Add all tasks with dependencies"
        ],
        task_generation="5 tasks created with dependency chain",
        completion="User reviews tasks, then Auto Zen implements"
    ),
    
    PlanUpdateFlow(
        flow_id="update-003",
        trigger="User fixes typo in task description",
        detection="File watcher detects tasks.json modified",
        analysis=[
            "Compare versions",
            "Changed field: tasks[3].description",
            "Impact: Documentation only, no code impact",
            "Version increment: PATCH (2.0.0 → 2.0.1)",
            "No code changes needed"
        ],
        sync_actions=[
            "Update metadata.json with new version",
            "Add changelog entry",
            "No tasks generated (documentation change)"
        ],
        task_generation="0 tasks (no code impact)",
        completion="Auto-saved, no sync required"
    ),
    
    PlanUpdateFlow(
        flow_id="update-004",
        trigger="User adds new page to design",
        detection="File watcher detects plan.json modified",
        analysis=[
            "Compare versions",
            "New field added: pages.reportsPage",
            "Impact: New page creation",
            "Version increment: MINOR (2.0.1 → 2.1.0)",
            "Existing code not affected (additive change)"
        ],
        sync_actions=[
            "Generate task: Create ReportsPage.vue component",
            "Generate task: Add route for /reports",
            "Generate task: Add navigation link",
            "Generate task: Design reports layout",
            "Generate task: Write tests for reports page",
            "Add tasks to queue"
        ],
        task_generation="5 tasks created",
        completion="Auto Zen implements new page"
    )
]

print("\nPLAN UPDATE FLOWS:")
print("-" * 130)

for flow in update_flows:
    print(f"\n[{flow.flow_id}] {flow.trigger}")
    print(f"  Detection: {flow.detection}")
    print(f"  Analysis Steps: {len(flow.analysis)}")
    for step in flow.analysis[:3]:
        print(f"    - {step}")
    print(f"  Sync Actions: {len(flow.sync_actions)}")
    print(f"  Tasks Generated: {flow.task_generation}")
    print(f"  Completion: {flow.completion}")

print("\n\n" + "=" * 130)
print("CODE SYNCHRONIZATION PIPELINE")
print("=" * 130)

sync_pipeline = """
PLAN CHANGE DETECTED
    ↓
[1] DETECT CHANGE
    ├─ File watcher triggers on plan.json save
    ├─ Read new plan.json
    ├─ Load previous version from metadata
    └─ Calculate diff
    ↓
[2] ANALYZE IMPACT
    ├─ What changed? (field-level diff)
    ├─ What's the impact? (code, tests, docs)
    ├─ What version increment? (major/minor/patch)
    ├─ Is code sync needed? (yes/no)
    └─ What files are affected?
    ↓
[3] GENERATE TASKS (if sync needed)
    ├─ For each affected component:
    │   ├─ Create update task
    │   ├─ Add acceptance criteria
    │   ├─ Link to plan version
    │   └─ Set dependencies
    ├─ Add test tasks
    ├─ Add validation tasks
    └─ Write to _ZENTASKS/tasks.json
    ↓
[4] UPDATE METADATA
    ├─ Increment version
    ├─ Add changelog entry
    ├─ Mark as "sync pending"
    ├─ Set lastUpdatedAt timestamp
    └─ Save metadata.json
    ↓
[5] NOTIFY USER
    ├─ Show notification:
    │   "Plan updated to v2.0.0"
    │   "5 tasks generated for code sync"
    │   "Review tasks before Auto Zen starts?"
    ├─ Options:
    │   [Review Tasks] [Start Auto Zen] [Cancel]
    └─ Log to activity feed
    ↓
[6] USER REVIEWS (optional)
    ├─ User can review generated tasks
    ├─ Adjust priorities
    ├─ Add/remove tasks
    ├─ Modify acceptance criteria
    └─ Approve for execution
    ↓
[7] AUTO ZEN EXECUTES
    ├─ Load tasks from _ZENTASKS/
    ├─ Execute in dependency order
    ├─ Mark tasks as done
    ├─ Run tests
    └─ Validate against plan
    ↓
[8] SYNC COMPLETE
    ├─ All tasks done
    ├─ Update metadata: "sync completed"
    ├─ Set lastSyncedAt timestamp
    ├─ Notify user: "Code synced with plan v2.0.0"
    └─ Generate completion report
    ↓
CODE MATCHES PLAN ✓
"""

print(sync_pipeline)
````

# Section 11: MCP Server & Coding Agent Integration (Enhanced)

## Overview: Comprehensive Agent Reporting & Autonomous Tracking

The MCP (Model Context Protocol) server now provides **comprehensive reporting, observation tracking, and verification workflows** that enable:

1. **Rich task status reporting** (not just done/not-done)
2. **Observation discovery** (finding new work autonomously)
3. **Automated verification tasks** (ensuring completion is real)
4. **Test failure tracking & recovery** (catch regressions immediately)
5. **Autonomous background tracking** (visual dashboard shows everything)

**Key principle**: Agents report *everything* they discover. The system tracks observations, generates verification tasks, and maintains visibility through a real-time dashboard.

---

## Task Design System: Super-Detailed Design Prompts

Every task includes a **super-detailed design prompt** that contains all necessary context:

```yaml
Task Structure:
  ├─ taskId & title (clear identification)
  ├─ superDetailedPrompt:
  │   ├─ description (WHAT)
  │   ├─ context (WHY)
  │   ├─ requirements (MUST DO)
  │   ├─ designReferences (design system specs, colors, fonts, components)
  │   ├─ files (what to read, what to write, what references)
  │   ├─ acceptanceCriteria (HOW TO VERIFY DONE)
  │   ├─ estimatedHours (WHEN)
  │   ├─ complexityLevel (easy|medium|hard|expert)
  │   └─ skillsRequired (list of skills needed)
  │
  ├─ dependencies (what must complete first)
  ├─ planReference (which plan section, which version)
  └─ relatedTasks (context for related work)

Example Task Prompt for Color System:
  
  WHAT: Create CSS variables for color palette
  WHY: Foundation for theming - all components depend on this
  MUST DO: 
    - Define 12 primary colors with 3 variants each
    - Support light AND dark theme switching
    - Meet WCAG AA contrast ratios
    - Colorblind-safe (no pure red/green)
  
  DESIGN REFS:
    - Color palette from design-system.json
    - Accessibility guide in plan
    - Breakpoints and theme switching notes
  
  FILES:
    - Read: Docs/Plans/my-app/design-system.json
    - Write: src/styles/colors.css, src/styles/_variables.scss
    - Reference: src/App.vue, tailwind.config.js
  
  VERIFY DONE:
    1. All 12 colors defined with 3 variants
    2. CSS variables use correct naming
    3. Light mode renders correctly
    4. Dark mode renders correctly
    5. All colors pass WCAG AA contrast test
    6. No pure red/green colors
    7. Can toggle theme in browser
```

## Verification Task Automation

When a task is marked done, the system **automatically creates verification tasks**:

```
TASK-001 (Color Palette) → reportTaskStatus(done)
    ↓
MCP Server processes
    ↓
Auto-creates: TASK-VERIFY-001 (Verify color palette works)
    ↓
Agent receives TASK-VERIFY-001 as next task
    ↓
Agent:
  1. Tests colors in actual components
  2. Checks theme toggle
  3. Verifies accessibility
  4. Reports via reportVerificationResult
    ↓
If verification passes:
  - Original task stays DONE ✓
  - Verification task DONE ✓
  - Next task in queue
    ↓
If verification fails:
  - Original task marked DONE_BUT_INCOMPLETE
  - Follow-up fix tasks created
  - Investigation tasks if tests broken
  - Back to queue with blockers
```

---

## Test Failure Recovery Protocol

```
1. Test fails during task execution
   └─ reportTestFailure called with failure details

2. MCP Server analyzes:
   - Was this test passing before?
   - Did code change cause it?
   - Or did test/dependencies change?

3. Creates investigation task:
   - TASK-INV-001: "FIX: Investigate test failure"
   - Priority: CRITICAL
   - Blocks original task

4. Investigation task details include:
   - Test name, failure message
   - Git diff analysis (what changed)
   - Suspected root causes
   - Investigation steps

5. Agent executes investigation:
   - Checks git history
   - Analyzes code changes
   - Runs diagnostic tests
   - Fixes root cause
   - Re-runs original test

6. Once test passes:
   - Investigation task marked done
   - Original task unblocked
   - Continues normal flow
```

# Section 11: MCP Server & Coding Agent Integration

## Overview: Enabling GitHub Copilot Without Zen Tasks Extension

When the Zen Tasks extension is **not available**, we provide an MCP (Model Context Protocol) server that allows GitHub Copilot or any coding agent to:

1. **Get next task** via `getNextTask` tool
2. **Report task completion** via `reportTaskDone` tool  
3. **Ask questions** about the plan via `askQuestion` tool

The plan files are stored in `Docs/Plans/` and read-only to the agent—ensuring all decisions are plan-driven.

---

```json
  "implementationNotes": "Changes made, decisions taken",
  "filesModified": ["src/styles/colors.css", "src/variables.ts"],
  "testsAdded": true,
  "testsPassed": true,
  "followUpTasks": ["optional - suggest new tasks discovered"]
}
```

**Output**:
```json
{
  "success": true,
  "taskId": "TASK-001",
  "status": "done",
  "message": "Task marked complete. 33 tasks remaining.",
  "nextTaskId": "TASK-002",
  "nextTaskPreview": {
    "title": "Build page navigation component",
    "priority": "high"
  }
}
```

## Coding Agent Definition

### .github/agents/Coding Agent.agent.md

```yaml
---
name: Coding Agent
description: Simple, task-driven coding agent using MCP tools to stay aligned with plan
argument-hint: Work on your next task. Use getNextTask, implement, then reportTaskDone.
tools: 
  - 'mcp/getNextTask'
  - 'mcp/reportTaskDone'
  - 'mcp/askQuestion'
  - 'vscode'
  - 'execute'
  - 'read'
  - 'edit'
  - 'search'

handoffs:
  - label: Get Next Task
    prompt: |
      Use mcp/getNextTask to retrieve the next available task.
      Read the task description, acceptance criteria, and plan reference.
      Implement the task following the acceptance criteria and plan specs.
      When done, use mcp/reportTaskDone to mark it complete.
      Suggest any follow-up tasks discovered.
    
  - label: Ask Planning Question
    prompt: |
      Use mcp/askQuestion to clarify any ambiguity.
      Provide the current task context and what you're unsure about.
      Follow the guidance from the plan.
      If still unclear, ask again with more specific context.
  
  - label: Hand to Auto Zen (Complex Work)
    agent: Auto Zen
    prompt: |
      This is a complex task that requires autonomous execution.
      Pass to Auto Zen with the current task context.
      Auto Zen will handle multi-step implementation, testing, and follow-up tasks.
  
  - label: Ask User for Clarification
    showContinueOn: true
    send: true
    prompt: |
      If the plan doesn't answer your question, ask the user directly.
      Explain what you need clarified and why.
      Wait for user input before proceeding.

# Coding Agent

## Purpose

Simple, focused task-driven coding agent that:
- Gets tasks from MCP server (not file system)
- Implements each task following plan guidance
- Asks questions to clarify ambiguities
- Reports completion and suggests follow-ups
- Stays aligned with the project plan at all times

## Workflow

1. **Get Task**: mcp/getNextTask → receives task with acceptance criteria
2. **Ask if Needed**: mcp/askQuestion → clarify requirements from plan
3. **Implement**: Write code matching plan specs and acceptance criteria
4. **Test**: Verify acceptance criteria met, add tests
5. **Report**: mcp/reportTaskDone → mark done, suggest follow-ups
6. **Repeat**: Go to step 1 for next task

## Key Principles

- Tasks come from MCP, not file system (keeps workspace clean)
- Always consult the plan when in doubt (mcp/askQuestion)
- Acceptance criteria are law (must all pass)
- If blocked, report it (mcp/reportTaskDone with status=blocked)
- If discover new work, suggest it in reportTaskDone
```

---

## Task Queue Management (In Memory)

Tasks are stored **in memory only**, never visible in the file system:

```
Memory Queue:
├─ TASK-001: Setup project (ready)
├─ TASK-002: Color palette (ready)  
├─ TASK-003: Navigation (blocked - waiting for colors)
├─ TASK-004: Task display (pending)
└─ ...33 more tasks

NOT in workspace directories:
✗ No _ZENTASKS/ folder
✗ No tasks.json file
✗ No TASK-*.md files

Accessed ONLY via:
✓ mcp/getNextTask
✓ mcp/reportTaskDone
✓ mcp/askQuestion
```

---

## Plan Storage (Read-Only)

Plans are visible and accessible to the agent:

```
Docs/Plans/my-app/
├─ plan.json                    ← Design choices, config
├─ metadata.json               ← Version history
├─ design-system.json          ← Visual specifications
├─ plan.md                     ← Human-readable spec
└─ tasks.json                  ← Generated task decomposition

Agent can:
✓ Read plan.json
✓ Look up design-system.json
✓ Reference plan.md
✓ Check acceptance criteria

Agent cannot:
✗ Modify plan files
✗ Change task definitions
✗ Edit metadata or version
```

## MCP Server Endpoints

### HTTP REST API (if using HTTP transport)

```
GET /mcp/nextTask?filter=ready&priority=high
  → Returns next task

POST /mcp/reportTaskDone
  → Request: { taskId, status, notes, files, tests }
  → Response: { success, nextTaskId, nextTaskPreview }

POST /mcp/askQuestion
  → Request: { question, context, taskId }
  → Response: { answer, evidence, guidance, relatedChoices }
```

### Stdio Protocol (if using JSON-RPC)

```json
Request:
{
  "jsonrpc": "2.0",
  "method": "getNextTask",
  "params": { "filter": "ready" },
  "id": 1
}

Response:
{
  "jsonrpc": "2.0",
  "result": { task: {...} },
  "id": 1
}
```

## Team 1: Planning Team

**Role**: Master planner and project tracker

**Responsibilities**:
- Takes the user-created plan from Planner Mode
- Decomposes plan into complete task tree
- Maintains overall project tracking
- Watches for plan updates and adjusts tasks
- Coordinates with Task Decomposition agent

**Agent Profile** (`config/agents/planning-team.yaml`):
```yaml
version: 1
name: "Planning Team"
role: "project_planner"
description: "Master planner that converts plans to tasks and tracks overall project"

goals:
  - "Convert user plans into comprehensive task trees"
  - "Maintain dependency-aware task graphs"
  - "Track overall project progress and health"
  - "Identify when plan updates require new tasks"
  - "Coordinate with Task Decomposition for complex work"

anti_goals:
  - "Never implement code directly"
  - "Never modify plan without user approval"
  - "Never create tasks not grounded in plan"

tool_permissions:
  read_files: true
  write_files: true  # For task files only
  modify_tasks: true
  access_network: false
  run_commands: false

execution_constraints:
  require_plan_before_action: true
  require_context_review: true
  max_parallel_actions: 5

prompt_templates:
  system: |
    You are the Planning Team agent. Your job is to understand the user's plan
    and break it into a complete, dependency-aware task tree.
    
    Read from: Docs/Plans/{plan-id}/plan.json
    Write to: _ZENTASKS/tasks.json and individual TASK-*.md files
    
    When plan updates, analyze impact and create appropriate tasks.
    Coordinate with Task Decomposition agent for complex tasks.
  
  planning: |
    Analyze the plan at {{planPath}}.
    Create tasks for all features/requirements.
    Ensure dependencies are correct.
    Mark ready tasks with no blockers.
```

## Team 2: Answer Team

**Role**: Context-aware Q&A system for current task

**Responsibilities**:
- Reads entire codebase (semantic understanding)
- Reads current plan
- Knows which task is currently being worked on
- Answers questions specifically related to current task context
- Provides plan references, code examples, design specs

**Agent Profile** (`config/agents/answer-team.yaml`):
```yaml
version: 1
name: "Answer Team"
role: "context_qa"
description: "Context-aware Q&A system that knows current task and plan"

goals:
  - "Answer questions related to current task"
  - "Provide relevant plan sections for reference"
  - "Show code examples from codebase"
  - "Clarify design decisions from plan"
  - "Guide coding agent with accurate context"

anti_goals:
  - "Never implement code"
  - "Never modify tasks or plan"
  - "Never answer questions outside current task scope"
  - "Never guess - admit when uncertain"

tool_permissions:
  read_files: true
  write_files: false
  modify_tasks: false
  access_network: false

execution_constraints:
  require_context_review: true
  max_depth: 3  # Can follow references but don't go too deep

prompt_templates:
  system: |
    You are the Answer Team agent. You have complete knowledge of:
    - The current plan (Docs/Plans/{{planId}})
    - The entire codebase (indexed semantically)
    - The current task being worked on ({{currentTaskId}})
    
    When asked a question:
    1. Check if it relates to current task
    2. Search plan for relevant design decisions
    3. Search codebase for existing patterns
    4. Provide clear, specific answer with references
    
    If uncertain, say so. Never guess.
  
  answer: |
    Current task: {{currentTask.title}}
    Question: {{question}}
    
    Search plan for relevant sections.
    Search code for similar implementations.
    Provide answer with references to plan + code.
```

## Team 3: Task Decomposition Agent

**Role**: Complexity watchdog and subtask creator

**Responsibilities**:
- Monitors tasks for complexity
- Detects when task is too large (>60 min estimate)
- Breaks tasks into 5-20 minute chunks (subtasks)
- Tells Planning Team to create proper subtasks
- Ensures granular, achievable work items

**Agent Profile** (`config/agents/task-decomposition.yaml`):
```yaml
version: 1
name: "Task Decomposition Agent"
role: "complexity_analyzer"
description: "Watches for complex tasks and breaks them into microtasks"

goals:
  - "Detect tasks estimated >60 minutes"
  - "Break complex tasks into 5-20 minute subtasks"
  - "Ensure subtasks have clear acceptance criteria"
  - "Notify Planning Team to create formal subtasks"
  - "Track subtask completion toward parent"

anti_goals:
  - "Never implement code"
  - "Never create subtasks <5 minutes (too granular)"
  - "Never create subtasks >20 minutes (still too complex)"

tool_permissions:
  read_files: true
  write_files: true
  modify_tasks: true

execution_constraints:
  require_plan_before_action: true
  max_parallel_actions: 1  # One task at a time

prompt_templates:
  system: |
    You are the Task Decomposition Agent. You watch for complex tasks
    and break them into microtasks (5-20 minute chunks).
    
    When you detect a task >60 minutes:
    1. Analyze what it requires
    2. Break into logical subtasks (5-20 min each)
    3. Assign dependencies between subtasks
    4. Tell Planning Team to create them formally
  
  decompose: |
    Task: {{taskTitle}}
    Estimated time: {{estimatedHours}} hours
    
    This is too complex. Break it into subtasks:
    - Each subtask: 5-20 minutes
    - Each subtask: One clear outcome
    - Each subtask: Testable/verifiable
    
    Create subtask breakdown and notify Planning Team.
```

## Team 4: Verification Team

**Role**: Automated + visual task verification

**Responsibilities**:
- **Auto verification**: Check code-based acceptance criteria
- **Visual verification**: Coordinate user-assisted UI testing
- **Test running**: Execute tests to verify completion
- **Regression detection**: Catch broken tests
- **User walkthrough**: Guide visual verification process

**Agent Profile** (`config/agents/verification-team.yaml`):
```yaml
version: 1
name: "Verification Team"
role: "task_verifier"
description: "Verifies task completion via automated and visual testing"

goals:
  - "Verify all acceptance criteria met"
  - "Run automated tests and report results"
  - "Coordinate visual verification with user"
  - "Detect regressions in existing tests"
  - "Guide user through visual testing process"

anti_goals:
  - "Never approve task without ALL criteria passing"
  - "Never skip visual verification when required"
  - "Never assume tests pass without running them"

tool_permissions:
  read_files: true
  write_files: false
  run_commands: true  # To run tests, start server
  modify_tasks: true  # To mark verified/failed

execution_constraints:
  require_tests_for_changes: true
  require_explicit_confirmation_for_commands: false  # Can start server

prompt_templates:
  system: |
    You are the Verification Team agent. When a task is reported done:
    
    1. Check acceptance criteria automatically
    2. Run tests related to the task
    3. If visual verification needed:
       - Start server/app
       - Create visual verification task
       - Wait for user to click "Ready"
       - Guide user through testing
    4. Report verification result
  
  verify: |
    Task: {{taskTitle}}
    Status: Reported as done
    
    Step 1: Check acceptance criteria
    {{#each acceptanceCriteria}}
      - {{this}}: [CHECK]
    {{/each}}
    
    Step 2: Run tests
    {{testCommand}}
    
    Step 3: Visual verification (if needed)
    - Start: {{startCommand}}
    - Test: {{visualTestChecklist}}
    - User confirms: [WAITING]
```

```
    ↓

    ↓

    ↓
```


When user clicks "Go Through Planning Process":

```

   └─ Shows current vs. proposed value
```

# Section 12: Copilot Orchestration Extension (COE) - Project Overview

## Project Vision

**Copilot Orchestration Extension (COE)** is a unified planning, task management, and multi-agent orchestration system for VS Code that enables developers to:

1. **Plan with confidence** - Guided, interactive planning interface with real-time visual feedback
2. **Execute autonomously** - Auto Zen, Zen Planner, and Coding agents operate independently with scoped context
3. **Stay synchronized** - Plans drive all execution; code changes sync back to plans
4. **Scale safely** - Dependency-aware task graphs prevent circular deadlocks and ensure correct sequencing
5. **Integrate anywhere** - MCP server enables GitHub Copilot even without Zen Tasks extension

---

## Problem Solved

Modern development requires:
- **Clear planning before coding** (avoid building wrong thing)
- **Task tracking that doesn't distract** (kanban boards are overhead)
- **Autonomous agents with guardrails** (fast execution + human oversight)
- **AI that understands project intent** (not just code snippets)

**COE solves this** by making plans the single source of truth that drives all agent behavior.

---

## Execution Flow: From Plan to Code

```
1. USER CREATES PLAN (Interactive Interface)
   ↓
2. Plan stored in Docs/Plans/{plan-id}/
   ├─ plan.json (structured data)
   ├─ design-system.json
   ├─ metadata.json (version info)
   └─ plan.md (human readable)
   ↓
3. DECOMPOSE Plan to Tasks
   ├─ Create task graph
   ├─ Detect dependencies
   ├─ Assign priorities
   └─ Validate DAG (no cycles)
   ↓
4. LOAD into Task Queue (In Memory)
   ├─ Ready tasks (no dependencies)
   ├─ Blocked tasks (waiting)
   └─ Completed tasks
   ↓
5. DISPATCH to Agent
   ├─ Auto Zen (complex, multi-step)
   ├─ Zen Planner (planning/requirements)
   ├─ Coding Agent (simple, MCP-driven)
   └─ Custom Agents (extensible)
   ↓
6. AGENT EXECUTES (with Plan Reference)
   ├─ Read task & acceptance criteria
   ├─ consult plan via askQuestion tool
   ├─ Implement code
   ├─ Test against acceptance criteria
   └─ Report completion
   ↓
7. TASK MARKED DONE
   ├─ Files modified tracked
   ├─ Follow-up tasks suggested
   ├─ Code changes synced to plan
   └─ Velocity metrics updated
   ↓
8. NEXT TASK RETRIEVED (Loop to step 5)
```

## Key Differentiators

| Aspect | Traditional | COE |
|--------|-------------|-----|
| **Planning** | Kanban board → vague cards | Structured plan → tasks auto-generated |
| **Agents** | Black box + generic prompts | Plan-aware with guided context |
| **Sync** | Manual copy-paste | Automatic code ↔ plan sync |
| **Dependencies** | Manual tracking | Automatic DAG with cycle detection |
| **Execution** | Human decides what's next | Queue-driven → agents pull work |
| **Accountability** | Hope agents stay on track | Plan + criteria = clear pass/fail |

# Section 11.6: MCP Tool → Agent Team Triggering (with Diagrams)

The coding agent now uses MCP tools to orchestrate specialized teams automatically.

**Trigger Matrix (conceptual):**
- `getNextTask` → fetches task + assigns Answer Team context
- `reportTaskStatus` → routes to Verification Team (auto + visual) and Task Decomposition (if too big)
- `reportObservation` → forwards to Planning Team + creates new tasks
- `reportTestFailure` → opens Investigation + blocks current task
- `reportVerificationResult` → closes verification or spawns follow-ups
- `askQuestion` → Answer Team pulls plan/code context

```mermaid
flowchart LR
    A[getNextTask] -->|assign current task| B(Answer Team)
    A -->|task size check| C(Task Decomposition)
    D[reportTaskStatus] -->|status=done| E(Verification Team)
    D -->|status=blocked/failed| F(Planning Team)
    D -->|partial| C
    G[reportObservation] --> F
    G --> H(New Task Queue)
    I[reportTestFailure] --> J(Investigation Task)
    J --> F
    K[reportVerificationResult] -->|passed| L(Close Task)
    K -->|failed/partial| H
    M[askQuestion] --> B
    E -->|needs visual| N(Visual Verification Flow)
```

# Visual Verification Panel Wiring (Backend Hooks)


**UI is scaffolded; now wire to real actions:**


- `startServer/restartServer/stopServer` → call backend/MCP to manage dev server; stream status via WebSocket.
- `markReady` → transition verification task from waiting → active; notify Verification Team.
- Checklist select → update verification state; on `passed` all, allow completion.
- `submitIssues` → create investigation/follow-up tasks (severity captured).
- `submitChangeRequest` → launch Plan Adjustment Wizard with proposed change + impact.
- Pull real data: taskId, planVersion, checklist, alreadyTested, retestRequired, notInScope, planHighlights.

```mermaid
sequenceDiagram

    VV->>MCP: submitChangeRequest(summary, impact)
    MCP-->>VT: verification updates
    MCP-->>PT: change requests → plan wizard
    MCP-->>VV: status events (WebSocket)
```

# Plan Adjustment Wizard (Flow)


**When user requests a change during verification:**


1) Capture proposed change (summary + optional impact).
2) Load current plan section; compute diff vs proposed.
3) Run scoped planning Q&A (reuse Section 9 prompts; only relevant questions).
4) Show impact analysis: version bump, affected components, new tasks, est. time.
5) On confirm: update plan (plan.json + metadata), bump version, regenerate tasks, enqueue new verifications.
6) Emit events to dashboard; close or requeue verification task.


```mermaid
flowchart LR
  CR[Change Request] --> D[Diff Current vs Proposed]
  D --> Q[Scoped Planning Questions]
  Q --> I[Impact Analysis]
  I --> U[Update Plan + Version]
  U --> T[Regenerate Tasks]
  T --> V[New Verifications]
  V --> WS[Emit Events to UI]
```

# MCP Endpoints & Payloads (Concrete Spec)


**Core HTTP/JSON-RPC shapes**
- `/mcp/nextTask` → {filter, priority} → returns task with superDetailedPrompt, acceptanceCriteria, planReference
- `/mcp/reportTaskStatus` → status + tests + observations → emits events; may spawn verification/investigation tasks
- `/mcp/reportObservation` → logs discovery/issue; optionally creates new task; emits dashboard alert
- `/mcp/reportTestFailure` → blocks task; creates investigation task; raises critical alert
- `/mcp/reportVerificationResult` → passes/fails verification; may create follow-ups
- `/mcp/askQuestion` → answers from plan/code context with confidence + evidence

# Audit Dashboard Spec (Plan vs Implementation)


**Panels:**
- Summary: total/complete/blocked/verification/investigation counts; plan version; velocity trend.
- Compliance: acceptance criteria coverage, verification pass rate, test pass rate.
- Drift: plan changes pending sync; code changes needing plan updates.
- Observations: open observations, resolved, SLA.
- Alerts: test failures, blocked tasks, visual verification pending.

```mermaid
graph TD
  Audit[Audit Dashboard]

  Audit --> Compliance[Compliance Metrics]
  Audit --> Drift[Plan-Code Drift]
  Audit --> Observations[Observations]
  Audit --> Alerts[Alerts]
```

**Data Sources:** WebSocket audit events + MCP task queue + plan metadata.


# Code ↔ Plan Sync & Drift Detection


- Watch plan files (plan.json, design-system.json, metadata.json) → trigger task generation.
- Watch code changes touching plan-mapped files → suggest plan updates (reverse mapping).
- Compute drift score: #plan changes unsynced + #code changes impacting plan sections.
- On drift > threshold → pause auto-exec, prompt Planning Team + Plan Adjustment Wizard.

```mermaid
flowchart LR
  PlanChange[Plan change] --> GenTasks[Generate/Update Tasks]
  CodeChange[Code change] --> Impact[Map to plan sections]
  Impact --> Suggest[Suggest plan update]
  Drift[Drift score] -->|>threshold| Pause[Pause auto-loop]
  Pause --> Wizard[Plan Adjustment Wizard]
```


# MCP Backend Handlers (TypeScript Pseudocode)


```ts
// getNextTask
router.get('/mcp/nextTask', async (req, res) => {
  const task = queue.next({ filter: req.query.filter, priority: req.query.priority });
  const plan = planReader.forTask(task.planId);
  res.json({
    success: true,
    task: enrichWithPrompt(task, plan),
    queueLength: queue.length(),
    nextTasksPreview: queue.peek(2)
  });
});

// reportTaskStatus
router.post('/mcp/reportTaskStatus', async (req, res) => {
  const result = taskService.updateStatus(req.body);
  const events = eventBuilder.fromTaskStatus(result);
  ws.broadcast(events);
  res.json({ success: true, ...result });
});

// reportObservation
router.post('/mcp/reportObservation', async (req, res) => {
  const obs = observationService.log(req.body);
  const maybeTask = taskService.maybeCreateFromObservation(obs);
  ws.broadcast(eventBuilder.observation(obs, maybeTask));
  res.json({ success: true, observationId: obs.id, newTaskCreated: maybeTask });
});

// reportTestFailure
router.post('/mcp/reportTestFailure', async (req, res) => {
  const inv = investigationService.create(req.body);
  taskService.block(req.body.taskId, inv.id);
  ws.broadcast(eventBuilder.testFailure(inv));
  res.json({ success: true, investigationTask: inv });
});

// reportVerificationResult
router.post('/mcp/reportVerificationResult', async (req, res) => {
  const vr = verificationService.apply(req.body);
  ws.broadcast(eventBuilder.verification(vr));
  res.json({ success: true, ...vr });
});

// askQuestion
router.post('/mcp/askQuestion', async (req, res) => {
  const answer = qaService.answer(req.body.question, req.body.context, req.body.currentTaskId);
  res.json(answer);
});
```


# Event Schema Typings & Subscription Flow


```ts
// events.ts
export type EventType = 'task-status' | 'observation' | 'verification' | 'test-failure' | 'server-status' | 'audit';

export interface BaseEvent { type: EventType; timestamp: string; }
export interface TaskStatusEvent extends BaseEvent {
  taskId: string;
  status: 'ready' | 'in-progress' | 'done' | 'blocked' | 'failed' | 'verification' | 'investigation';
  progress?: number;
  message?: string;
}
export interface VerificationEvent extends BaseEvent {
  taskId: string;
  originalTaskId: string;
  checklist: { id: string; status: string; note?: string }[];
  requiresUserReady: boolean;
  serverStatus: 'running' | 'stopped' | 'starting' | 'error';
  planHighlights?: string[];
}
export interface AuditEvent extends BaseEvent {
  counts: { total: number; completed: number; blocked: number; verification: number; investigation: number };
  planVersion: string;
  velocity?: { tasksPerDay: number; trend: 'up' | 'flat' | 'down' };
}

export type OrchestratorEvent = TaskStatusEvent | VerificationEvent | AuditEvent | BaseEvent;
```

**Subscription flow (panels):**
1) Each panel opens WebSocket → subscribes to all events.
2) Filters by type: Status/Orchestrator panels use task-status & audit; Visual Verification uses verification & server-status; Alerts use test-failure & observation.
3) Debounce UI updates; keep last N events for history.
4) Reconnect with backoff on disconnect.


# Plan Adjustment Wizard: Question Set & Version Rules


**Scoped questions (reuse Section 9, filtered by impacted area):**
- Color/theme change → ask: palette source, contrast targets, dark mode rules, affected components.
- Layout change → ask: breakpoints, navigation behavior, key pages impacted.
- API change → ask: contract, consumers, migrations, backward compatibility.
- Copy/UX change → ask: locales, tone, screens affected, acceptance criteria updates.

**Version bump rules:**
- MAJOR: Breaking API/UX, redesign, incompatible data.
- MINOR: New features, additive visuals, non-breaking options.
- PATCH: Fixes, copy tweaks, small accessibility fixes.

**Flow:**
1) Detect impacted plan sections from change request.
2) Ask only relevant questions; prefill current answers; user edits.
3) Compute version bump (major/minor/patch) and impacted tasks.
4) Regenerate tasks + verification items; enqueue.
5) Emit audit/drift reset.
